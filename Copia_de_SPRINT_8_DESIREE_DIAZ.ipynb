Validacions bàsiques de les dades i decisions d'eliminació
🎯 Objectiu: assegurar que les dades compleixen unes condicions mínimes de qualitat i determinar quines files poden continuar sent analitzades i quines s'han de descartar.
En aquest exercici hauràs de validar aspectes clau del dataset que són crítics en qualsevol context real d'anàlisi de dades, prenent decisions justificades sobre la validesa o eliminació de registres.

Aspectes a validar:

» Existeixen identificadors únics (DNI) per a cada persona?
» Què faràs amb les files duplicades o registres repetits?
» Hi ha valors obligatoris que falten o informació incoherent?
» Detectes formats incorrectes, errors de codificació o valors impossibles?

```python
# IDENTIFICAMO SPROBLEMAS DE CODIFICACIàON:

# Crear una copia del DataFrame original
df_limpio = df.copy()

# Mostrar los nombres de las columnas
print("NOMBRES DE LAS COLUMNAS:")
print(df_limpio.columns.tolist())

# Mostrar las primeras cinco filas
print("\nPRIMERAS CINCO FILAS:")
display(df_limpio.head())# S
```

```python
# PASO 1. CORREGIR ERRORES DE CODIFICACIÓN

def corregir_texto(texto):

    if isinstance(texto, str):
        try:
            return texto.encode("cp1252").decode("utf-8")
        except (UnicodeError, ValueError):
            return texto

    return texto


# Corregir los nombres de las columnas
df_limpio.columns = [
    corregir_texto(columna) for columna in df_limpio.columns
]

# Corregir los valores de texto
for columna in df_limpio.select_dtypes(include="object").columns:
    df_limpio[columna] = df_limpio[columna].map(corregir_texto)


# Mostrar los datos corregidos
display(df_limpio.head())
```

```python
# PASO 2. DETECTAR Y ELIMINAR FILAS DUPLICADAS EXACTAS

# Contar filas antes de eliminar duplicados
filas_antes = len(df_limpio)

# Eliminar filas duplicadas
df_limpio = df_limpio.drop_duplicates()

# Contar filas después
filas_despues = len(df_limpio)

# Mostrar cuántas filas se han eliminado
print("Filas antes:", filas_antes)
print("Filas después:", filas_despues)
print("Filas eliminadas:", filas_antes - filas_despues)
```

No hemos detectado filas duplicadas exactas de momento



```python
# PASO 3. COMPROBAR LA UNICIDAD DEL DNI

# Detectar registros con DNI duplicado
dni_duplicados = df_limpio[
    df_limpio["DNI"].notna() &
    df_limpio["DNI"].duplicated(keep=False)
]

# Mostrar los registros con DNI repetido
print("Registros con DNI duplicado:", len(dni_duplicados))

display(dni_duplicados.sort_values("DNI"))
```

```python
# PASO 3.1. ANALIZAR LOS DNI DUPLICADOS

# Seleccionar los datos importantes para comparar
columnas = [
    "DNI",
    "Nom",
    "Cognoms",
    "Dia de Naixement",
    "Mes de Naixement",
    "Any de Naixement",
    "Gènere",
    "Ciutat",
    "Salari mensual"
]

# Mostrar los registros ordenados por DNI
display(
    dni_duplicados[columnas].sort_values("DNI")
)
```


Comparamos algunas columnas como: los nombres, apellidos, fechas de nacimiento, género y demás información asociada a cada DNI.

Si dos registros tienen el mismo DNI pero corresponden aparentemente a personas diferentes, existe una inconsistencia.

Tenemos los 24 registros correspondientes a 8 DNI duplicados

Coinciden el nombre, los apellidos y la fecha de nacimiento, pero existen diferencias en la ciudad, el género y el salario. Comparamos  diferencias entre DNI , así podremos distinguir los registros realmente equivalentes de aquellos que tienen información contradictoria.

Tampoco podemos determinar con seguridad qué registro debemos conservar.

```python
# PASO 3.6. GUARDAR LOS DNI DUPLICADOS EN OTRA TABLA

# Crear una tabla con todos los registros de DNI repetidos
df_revision = df_limpio[
    df_limpio["DNI"].duplicated(keep=False)
    & df_limpio["DNI"].notna()
].copy()

# Mostrar la tabla ordenada por DNI
display(df_revision.sort_values("DNI"))
```

```python
# PASO 3.2. MOSTRAR TODOS LOS DNI DUPLICADOS; para poder tomar decisiones,

# Mostrar la tabla completa
with pd.option_context("display.max_rows", None):
    display(dni_duplicados[columnas].sort_values("DNI"))
```

Los casos más graves son los DNI 54013571W y 90585882A, porque aparecen asociados a personas con nombres y fechas de nacimiento diferentes. Sin una fuente externa fiable, no podemos determinar a quién pertenece realmente cada DNI.

En los demás casos hay coincidencias de identidad, pero debemos comprobar si las diferencias de salario o ciudad son errores, actualizaciones o registros distintos.

Además, que dos filas coincidan en las nueve columnas mostradas no significa que sean duplicados exactos, porque el dataset tiene 20 columnas.

```python
# PASO 3.5. SEPARAR LOS DNI CONFLICTIVOS

# DNI que presentan información contradictoria
dni_conflictivos = [
    "15909601K",
    "16618423V",
    "49529438G",
    "54013571W",
    "90585882A"
]

# Guardar los registros conflictivos en otra tabla
df_descartados = df_limpio[
    df_limpio["DNI"].isin(dni_conflictivos)
].copy()

# Separar los registros conflictivos del DataFrame de análisis
df_limpio = df_limpio[
    ~df_limpio["DNI"].isin(dni_conflictivos)
].copy()

# Mostrar resultados
print("Registros separados:", len(df_descartados))
print("Registros restantes:", len(df_limpio))

display(df_descartados)
```

Sabemos que hay cuatro DNI cuyos registros solo se diferenciaban en la columna Unnamed: 0.

Vamos a conservar un registro de cada uno y guardar los repetidos en df_descartados.

```python
# PASO 3.6. SEPARAR LOS DNI DUPLICADOS RESTANTES

# Identificar las repeticiones de DNI
duplicados = df_limpio["DNI"].notna() & df_limpio["DNI"].duplicated()

# Guardar las filas repetidas en la tabla de descartados
df_descartados = pd.concat([
    df_descartados,
    df_limpio[duplicados]
])

# Conservar un registro por DNI
df_limpio = df_limpio[~duplicados].copy()

# Mostrar resultados
print("Registros separados en total:", len(df_descartados))
print("Registros restantes:", len(df_limpio))
print("DNI duplicados restantes:",
      df_limpio["DNI"].dropna().duplicated().sum())
```

```python
# PASO 4. COMPROBAR VALORES NULOS EN CAMPOS CLAVE

# Seleccionar los campos clave
campos_clave = ["Nom", "Cognoms", "DNI"]

# Contar valores nulos
print("VALORES NULOS EN CAMPOS CLAVE:")
print(df_limpio[campos_clave].isnull().sum())

# Mostrar registros con algún campo clave vacío
registros_incompletos = df_limpio[
    df_limpio[campos_clave].isnull().any(axis=1)
]

display(registros_incompletos)
```


Como no tenemos información para identificar a la persona ni recuperar sus datos, está justificado excluir este registro del análisis.


```python
# PASO 4.1. SEPARAR REGISTROS IRRECUPERABLES

# Identificar filas sin información útil
columnas_datos = df_limpio.columns.drop("Unnamed: 0")

filas_vacias = df_limpio[columnas_datos].isnull().all(axis=1)

# Guardar los registros en la tabla de descartados
df_descartados = pd.concat([
    df_descartados,
    df_limpio[filas_vacias]
])

# Retirarlos del DataFrame de análisis
df_limpio = df_limpio[~filas_vacias].copy()

# Mostrar resultados
print("Registros irrecuperables:", filas_vacias.sum())
print("Total registros separados:", len(df_descartados))
print("Registros válidos restantes:", len(df_limpio))
```

```python
# PASO 3.2. REVISAR HIJOS SIN NÚMERO REGISTRADO

# Personas que tienen hijos, pero no consta cuántos
sin_numero = df_transformado[
    (df_transformado["Fills"] == True) &
    (df_transformado["Nombre_fills"].isna())
]

print("Personas con hijos, pero sin número registrado:", len(sin_numero))

display(sin_numero[["Fills", "No Fills", "Nombre_fills"]])
```

```python
# PASO 4. REVISAR TEMPERATURAS NULAS

print("Temperaturas nulas:",
      df_transformado["Temperatura_mitjana_ciutat"].isna().sum())

# Mostrar las ciudades de los registros sin temperatura
sin_temperatura = df_transformado[
    df_transformado["Temperatura_mitjana_ciutat"].isna()
]

print("\nCiudades con temperaturas faltantes:")
print(sin_temperatura["Ciutat"].value_counts())
```

Utilizaremos la temperatura media de cada ciudad. Si alguna ciudad no tiene ninguna temperatura registrada, utilizaremos la media general del dataset.

```python
# PASO 4.1. IMPUTAR TEMPERATURAS FALTANTES

# Calcular la temperatura media de cada ciudad
media_ciudad = df_transformado.groupby("Ciutat")["Temperatura_mitjana_ciutat"].transform("mean")

# Rellenar los valores nulos con la media de su ciudad
df_transformado["Temperatura_mitjana_ciutat"] = (
    df_transformado["Temperatura_mitjana_ciutat"].fillna(media_ciudad)
)

# Si todavía quedan nulos, utilizar la media general
media_general = df_transformado["Temperatura_mitjana_ciutat"].mean()

df_transformado["Temperatura_mitjana_ciutat"] = (
    df_transformado["Temperatura_mitjana_ciutat"].fillna(media_general)
)

# Comprobar resultados
print("Temperaturas nulas restantes:",
      df_transformado["Temperatura_mitjana_ciutat"].isna().sum())
```

Se han detectado 67 temperaturas nulas. Se imputan utilizando la temperatura media de cada ciudad. Cuando no hay datos suficientes de una ciudad, se utiliza la media general del dataset. Se conservan las temperaturas originales disponibles.

```python
# PASO 5. REVISAR LOS SALARIOS

# Comprobar salarios nulos
print("Salarios nulos:",
      df_transformado["Salari mensual"].isna().sum())

# Mostrar una muestra de los salarios
print("\nEjemplos de salarios:")
print(df_transformado["Salari mensual"].head(10))
```

```python
# PASO 5.1. NORMALIZAR LOS SALARIOS

# Recuperar los salarios originales
salarios = df_limpio["Salari mensual"].copy()

# Limpiar símbolos y separadores de miles
salarios = (
    salarios.astype("string")
    .str.lower()
    .str.replace("euros", "", regex=False)
    .str.replace("€", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.replace(".", "", regex=False)
    .str.strip()
)

# Convertir los salarios a números
df_transformado["Salari mensual"] = pd.to_numeric(
    salarios, errors="coerce"
)

# Comprobar resultados con punto de miles
print("Salarios nulos:",
      df_transformado["Salari mensual"].isna().sum())

display(
    df_transformado[["Salari mensual"]].head(10)
    .style.format({"Salari mensual": "{:,.0f} €"}, thousands=".", decimal=",")
)
```

```python
# PASO 5.2. REVISAR SALARIOS NO CONVERTIDOS

# Identificar los salarios que no se han podido convertir
salarios_revisar = df_transformado[
    df_transformado["Salari mensual"].isna()
]

# Mostrar los valores originales y los datos asociados
display(df_limpio.loc[
    salarios_revisar.index,
    ["Salari mensual", "Càrrec", "Grup Professional"]
])
```

```python
# PASO 5.3. RECUPERAR SALARIOS ESCRITOS EN TEXTO

# Convertir los salarios expresados en palabras
equivalencias = {
    "mil vuit-cents": 1800,
    "dos mil": 2000,
    "tres mil": 3000
}

# Recuperar los salarios originales
salarios_texto = df_limpio["Salari mensual"].str.strip().str.lower()

# Sustituir los valores que podemos recuperar
salarios_recuperados = salarios_texto.map(equivalencias)

# Completar únicamente los salarios que quedaron nulos
df_transformado["Salari mensual"] = (
    df_transformado["Salari mensual"].fillna(salarios_recuperados)
)

# Comprobar resultados
print("Salarios nulos restantes:",
      df_transformado["Salari mensual"].isna().sum())
```

Se han detectado salarios escritos en palabras y
valores no disponibles (n/d). Los salarios expresados en palabras se convierten
a números porque conocemos su importe exacto. Los valores n/d se mantienen como nulos para imputarlos posteriormente con un criterio analítico.

Utilizaremos la mediana salarial del mismo cargo y grupo profesional. Es un criterio razonable porque compara trabajadores con características laborales similares y reduce la influencia de salarios extremos.

```python
# PASO 5.4. IMPUTAR LOS SALARIOS FALTANTES

# Convertir los salarios a tipo decimal
df_transformado["Salari mensual"] = (
    df_transformado["Salari mensual"].astype("float64")
)

# Contar salarios nulos antes
print("Salarios nulos antes:",
      df_transformado["Salari mensual"].isna().sum())

# Calcular la mediana por cargo y grupo profesional
mediana_salario = df_transformado.groupby(
    ["Càrrec", "Grup Professional"]
)["Salari mensual"].transform("median")

# Completar los salarios faltantes
df_transformado["Salari mensual"] = (
    df_transformado["Salari mensual"].fillna(mediana_salario)
)

# Comprobar resultados
print("Salarios nulos después:",
      df_transformado["Salari mensual"].isna().sum())
```

Se han recuperado 22 salarios escritos en palabras, convirtiéndolos a valores numéricos.

Los 14 salarios faltantes se han imputado utilizando la mediana salarial de trabajadores con el mismo cargo y grupo profesional. Este criterio permite estimar los salarios a partir de empleados con características laborales similares y reduce la influencia de valores extremos.

Después de la imputación, no quedan salarios nulos en el dataset de análisis.

```python
# PASO 3. EXTRA - COMPARACIÓN SALARIAL EN PORCENTAJES

# Calcular la media salarial general
media_general = df_transformado["Salari mensual"].mean()

# Crear tabla de comparación
tabla_porcentajes = tabla_salarios[["Media"]].copy()

# Calcular diferencia porcentual respecto a la media general
tabla_porcentajes["Diferencia %"] = (
    (tabla_porcentajes["Media"] / media_general - 1) * 100
)

# Mostrar resultados
display(
    tabla_porcentajes.style.format({
        "Media": "{:,.2f} €",
        "Diferencia %": "{:+.2f} %"
    }, thousands=".", decimal=",")
)
```

```python
# PASO 4. EXTRA - TAMAÑO DE GRUPOS Y RATIOS SALARIALES

# Calcular número de personas y salario medio por género
tabla_ratios = df_transformado.groupby("Gènere")["Salari mensual"].agg(
    Personas="count",
    Media="mean"
)

# Calcular porcentaje de personas por género
tabla_ratios["Personas %"] = (
    tabla_ratios["Personas"] / tabla_ratios["Personas"].sum() * 100
)

# Calcular ratio salarial respecto a la media general
tabla_ratios["Ratio salarial"] = (
    tabla_ratios["Media"] / media_general
)

# Ordenar por salario medio
tabla_ratios = tabla_ratios.sort_values("Media", ascending=False)

# Mostrar resultados
display(
    tabla_ratios.style.format({
        "Media": "{:,.2f} €",
        "Personas %": "{:.2f} %",
        "Ratio salarial": "{:.2f}"
    }, thousands=".", decimal=",")
)
```

```python
# PASO 2.4 - NORMALIZAR NOMBRES DE CIUDADES /En el CSV aparece como A Coru\u00f1a (con los caracteres \u00f1 escritos literalmente).

df_enriquecido = df_transformado.copy()

# Corregir nombres para que coincidan con el CSV
df_enriquecido["Ciutat"] = df_enriquecido["Ciutat"].replace({
    "München": "Munchen",
    "A Coruña": r"A Coru\u00f1a"
})

# Comprobar las ciudades que siguen sin coincidencia
ciudades_pendientes = (
    set(df_enriquecido["Ciutat"].dropna())
    - set(df_ciudades["Ciutat"].dropna())
)

print("CIUDADES TODAVÍA SIN COINCIDENCIA:")
print(sorted(ciudades_pendientes))
```

```python
# ==========================================================
# PASO 3 - UNIR DATASET PRINCIPAL Y CIUDADES
# ==========================================================

df_enriquecido = df_enriquecido.merge(
    df_ciudades,
    left_on=["Ciutat", "País d'origen"],
    right_on=["Ciutat", "País"],
    how="left",
    validate="many_to_one",
    indicator=True
)

# Comprobar dimensiones
print("DIMENSIONES DEL DATASET ENRIQUECIDO:")
print(df_enriquecido.shape)

# Comprobar coincidencias
print("\nRESULTADO DE LA UNIÓN:")
print(df_enriquecido["_merge"].value_counts())

# Mostrar primeras filas
display(df_enriquecido.head())
```

aunque las columnas de nacimiento no tienen valores vacíos, existen fechas incorrectas que impiden calcular la edad de algunos empleados.. Utilizaremos las fechas válidas para calcular la edad y dejaremos el ajuste por edad como no calculable cuando la fecha sea incorrecta.
Así cumplimos el enunciado: identificar la información insuficiente, justificar nuestra decisión y explicar su impacto en el salario ajustado.

```python
# ==========================================================
# PASO 5 - CALCULAR EDADES Y FRANJAS DE EDAD / podemos calcular el ajuste por edad para 908 empleados. Los otros 77 conservarán el ajuste como desconocido.
# ==========================================================

# 1. Guardar las fechas de nacimiento
df_enriquecido["Fecha_nacimiento"] = fechas

# 2. Calcular la edad a fecha 18/10/2026
fecha_referencia = pd.Timestamp("2026-10-18")

df_enriquecido["Edad"] = (
    fecha_referencia.year
    - df_enriquecido["Fecha_nacimiento"].dt.year
    - (
        fecha_referencia.strftime("%m%d")
        < df_enriquecido["Fecha_nacimiento"].dt.strftime("%m%d")
    )
)

# 3. Descartar edades imposibles o inferiores a 18 años
df_enriquecido.loc[
    (df_enriquecido["Edad"] < 18) |
    (df_enriquecido["Edad"] > 100),
    "Edad"
] = float("nan")

# 4. Asignar franjas de edad
def clasificar_edad(edad):

    if pd.isna(edad):
        return None
    elif edad < 30:
        return "18-29"
    elif edad < 45:
        return "30-44"
    elif edad < 60:
        return "45-59"
    else:
        return "60+"

df_enriquecido["Franja_edad"] = (
    df_enriquecido["Edad"].apply(clasificar_edad)
)

# 5. Mostrar resultados
print("EMPLEADOS POR FRANJA DE EDAD:")
print(df_enriquecido["Franja_edad"].value_counts(dropna=False))

display(
    df_enriquecido[
        ["Nom", "Fecha_nacimiento", "Edad", "Franja_edad"]
    ].head(10)
)
```

```python
# ==========================================================
# PASO 6 - APLICAR INCREMENTOS SALARIALES
# ==========================================================

# 1. Incremento por grupo profesional
df_enriquecido["Incremento_grupo"] = (
    df_enriquecido["Grup Professional"].map(
        politica["base_increment_by_group"]
    )
)

# 2. Incremento por hijos
def incremento_hijos(numero_hijos):

    if pd.isna(numero_hijos):
        return None
    elif numero_hijos > 0:
        return politica["children_adjustment"]["has_children"]
    else:
        return politica["children_adjustment"]["no_children"]

df_enriquecido["Incremento_hijos"] = (
    df_enriquecido["Nombre_fills"].apply(incremento_hijos)
)

# 3. Incremento por edad
df_enriquecido["Incremento_edad"] = (
    df_enriquecido["Franja_edad"].map(
        politica["age_adjustment"]
    )
)

# 4. Comprobar resultados
print("VALORES VACÍOS EN LOS INCREMENTOS:")
print(
    df_enriquecido[
        ["Incremento_grupo", "Incremento_hijos", "Incremento_edad"]
    ].isna().sum()
)

display(
    df_enriquecido[
        ["Nom", "Grup Professional", "Incremento_grupo",
         "Nombre_fills", "Incremento_hijos",
         "Franja_edad", "Incremento_edad"]
    ].head(10)
)
```

Pero nuestro CSV contiene Cost_vida_index, un índice numérico, no las categorías low, medium y high.
Por tanto, antes de aplicar los porcentajes necesitamos saber cómo convertir ese índice en categorías. El JSON no especifica los límites, así que tendremos que definirlos y justificar nuestra decisión.

```python
# ==========================================================
# PASO 7.1 - ANALIZAR EL ÍNDICE DE COSTE DE VIDA
# ==========================================================

# 1. Estadísticas del índice
print("ESTADÍSTICAS DEL COSTE DE VIDA:")
print(df_ciudades["Cost_vida_index"].describe())

# 2. Mostrar ciudades e índices
print("\nCOSTE DE VIDA POR CIUDAD:")

display(
    df_ciudades[
        ["Ciutat", "País", "Cost_vida_index"]
    ].sort_values("Cost_vida_index")
)
```

```python
# PASO 7.2 - CLASIFICAR EL COSTE DE VIDA

def clasificar_coste(indice):

    if pd.isna(indice):
        return None
    elif indice < 68:
        return "low"
    elif indice <= 78:
        return "medium"
    else:
        return "high"

# Crear categoría de coste de vida
df_enriquecido["Categoria_coste"] = (
    df_enriquecido["Cost_vida_index"].apply(clasificar_coste)
)

# Aplicar incrementos del JSON
df_enriquecido["Incremento_coste"] = (
    df_enriquecido["Categoria_coste"].map(
        politica["cost_of_living_adjustment"]
    )
)

# Comprobar resultados
print("EMPLEADOS POR CATEGORÍA:")
print(df_enriquecido["Categoria_coste"].value_counts(dropna=False))

print("\nINCREMENTOS SIN CALCULAR:")
print(df_enriquecido["Incremento_coste"].isna().sum())
```

```python
# ==========================================================
# PASO 8 - CALCULAR EL SALARIO AJUSTADO
# ==========================================================

# 1. Seleccionar los cuatro incrementos
columnas_incrementos = [
    "Incremento_grupo",
    "Incremento_hijos",
    "Incremento_edad",
    "Incremento_coste"
]

# 2. Calcular el incremento total
df_enriquecido["Incremento_total"] = (
    df_enriquecido[columnas_incrementos].sum(
        axis=1, min_count=4
    )
)

# 3. Aplicar el límite máximo del 15 %
limite = politica["rules"]["max_total_increment_percent"]

df_enriquecido["Incremento_total"] = (
    df_enriquecido["Incremento_total"].clip(upper=limite)
)

# 4. Calcular el salario ajustado
df_enriquecido["Salario_ajustado"] = (
    df_enriquecido["Salari mensual"] *
    (1 + df_enriquecido["Incremento_total"] / 100)
)

# 5. Mostrar resultados
print("SALARIOS CALCULADOS:")
print(df_enriquecido["Salario_ajustado"].notna().sum())

print("\nSALARIOS PENDIENTES:")
print(df_enriquecido["Salario_ajustado"].isna().sum())

display(
    df_enriquecido[
        ["Nom", "Salari mensual",
         "Incremento_total", "Salario_ajustado"]
    ].head(10).style.format({
        "Salari mensual": "{:,.2f} €",
        "Incremento_total": "{:.2f} %",
        "Salario_ajustado": "{:,.2f} €"
    }, na_rep="—")
)
```

Se ha enriquecido el dataset principal de 985 empleados mediante la integración del archivo CSV de ciudades europeas y la política salarial definida en formato JSON.

```python
# ==========================================================
# SPRINT 8 - NIVEL 2 - EJERCICIO 2
# PASO 1 - SELECCIONAR COLUMNAS
# ==========================================================

# Columnas identificadoras
identificadores = [
    "DNI",
    "Nom",
    "Ciutat",
    "Grup Professional",
    "Gènere"
]

# Variables que transformaremos a formato largo
variables = [
    "Salari mensual",
    "Salario_ajustado",
    "Km_anuals",
    "Consum_mitja_L_100km",
    "Temperatura_mitjana_ciutat"
]

# Comprobar que las columnas existen
print("COLUMNAS IDENTIFICADORAS:")
print(identificadores)

print("\nVARIABLES MEDIDAS:")
print(variables)

# Mostrar una muestra del dataset
display(
    df_enriquecido[identificadores + variables].head()
)
```

```python

# PASO 2 - TRANSFORMAR A FORMATO LARGO
# ==========================================================

df_largo = df_enriquecido.melt(
    id_vars=identificadores,
    value_vars=variables,
    var_name="Variable",
    value_name="Valor"
)

# Comprobar dimensiones
print("DIMENSIONES DEL DATASET ORIGINAL:")
print(df_enriquecido.shape)

print("\nDIMENSIONES DEL DATASET LARGO:")
print(df_largo.shape)

# Mostrar las primeras filas
display(df_largo.head(10))
```

```python

# PASO 3 - COMPROBAR VALORES NULOS
# ==========================================================

# Contar valores nulos por variable
nulos = df_largo.groupby("Variable")["Valor"].apply(
    lambda x: x.isna().sum()
)

print("VALORES NULOS POR VARIABLE:")
print(nulos)

# Comprobar número de registros por variable
print("\nREGISTROS POR VARIABLE:")
print(df_largo["Variable"].value_counts())
```

```python

# PASO 4 - ESTADÍSTICAS AGRUPADAS
# ==========================================================

# Calcular la media de cada variable por grupo profesional
estadisticas = df_largo.groupby(
    ["Grup Professional", "Variable"]
)["Valor"].mean().round(2)

print("MEDIA POR GRUPO PROFESIONAL Y VARIABLE:")
display(estadisticas.reset_index())
```

```python
# ==========================================================
# SPRINT 8 - NIVEL 2 - EJERCICIO 2
# CONCLUSIONES FINALES
# ==========================================================

print("CONCLUSIONES DEL EJERCICIO 2")
print("=" * 50)

print("1. El dataset original tenía 985 filas y 33 columnas.")

print("2. Con melt() hemos obtenido 4925 filas y 7 columnas.")

print("3. El formato ancho facilita consultar todos los datos de un empleado.")

print("4. El formato largo facilita agrupar y comparar una misma variable.")

print("5. Con groupby() podemos calcular medias por grupo profesional.")

print("6. Los valores nulos se han conservado sin sustituirlos por cero.")

print("7. No debemos mezclar variables con unidades diferentes.")
```

```python

# PASO 1 - COMPROBAR COLUMNAS
# ==========================================================

print("COLUMNAS DISPONIBLES:")
print(df_enriquecido.columns.tolist())

# Comprobar los datos que utilizaremos
columnas = [
    "Salari mensual",
    "Nombre_fills",
    "Cost_vida_index",
    "Edad"
]

for columna in columnas:
    print(columna, ":", columna in df_enriquecido.columns)
```

```python

# PASO 2 - ÍNDICE DE DEPENDENCIA FAMILIAR
# ==========================================================

# Crear una copia del dataset original
df_derivadas = df_enriquecido.copy()

# Crear el índice de dependencia familiar
# Sumamos 1 para evitar dividir entre cero cuando no tiene hijos
df_derivadas["Indice_dependencia_familiar"] = (
    df_derivadas["Salari mensual"] /
    (df_derivadas["Nombre_fills"] + 1)
)

# Mostrar resultados con formato de euros/FALTA INDICE MOVILIDAD
display(
    df_derivadas[
        ["Nom", "Salari mensual", "Nombre_fills",
         "Indice_dependencia_familiar"]
    ].head(10).style.format({
        "Salari mensual": "{:,.0f} €",
        "Nombre_fills": "{:.0f}",
        "Indice_dependencia_familiar": "{:,.2f} €"
    }, na_rep="—")
)

# Comprobar casos límite del índice familiar

print("Número de hijos nulo:",
      df_derivadas["Nombre_fills"].isna().sum())

print("Número de hijos negativo:",
      (df_derivadas["Nombre_fills"] < 0).sum())

print("Índices familiares nulos:",
      df_derivadas["Indice_dependencia_familiar"].isna().sum())
```

Los 32 índices nulos corresponden a empleados cuyo número de hijos es desconocido. No debemos sustituir esos valores por cero, porque estaríamos suponiendo que no tienen hijos.

Además, no existen valores negativos en Nombre_fills, por lo que no hemos detectado esa incoherencia.

Se ha creado el índice de dependencia familiar dividiendo el salario mensual entre el número de hijos más uno. Se añade uno al denominador para evitar divisiones por cero cuando el empleado no tiene hijos.

Se han detectado 32 registros sin información sobre el número de hijos, que generan 32 índices nulos. Estos valores se mantienen como nulos para evitar realizar suposiciones sobre la situación familiar de los empleados.

No se han encontrado números negativos de hijos.

Este indicador permite explorar la relación entre salario y número de hijos, aunque no representa directamente la capacidad económica real de una familia.

```python

# PASO 3 - RATIO SALARIO / COSTE DE VIDA
# Comparar salarios según el coste de vida de cada ciudad
# ==========================================================

# Crear la segunda variable derivada
df_derivadas["Ratio_salario_coste_vida"] = (
    df_derivadas["Salari mensual"] /
    df_derivadas["Cost_vida_index"].replace(0, float("nan"))
)

# Mostrar resultados con formato
display(
    df_derivadas[
        ["Nom", "Ciutat", "Salari mensual",
         "Cost_vida_index", "Ratio_salario_coste_vida"]
    ].head(10).style.format({
        "Salari mensual": "{:,.0f} €",
        "Cost_vida_index": "{:.0f}",
        "Ratio_salario_coste_vida": "{:.2f}"
    }, na_rep="—")
)


# Comprobar casos límite del ratio salario / coste de vida

print("Costes de vida nulos:",
      df_derivadas["Cost_vida_index"].isna().sum())

print("Costes de vida iguales a cero:",
      (df_derivadas["Cost_vida_index"] == 0).sum())

print("Costes de vida negativos:",
      (df_derivadas["Cost_vida_index"] < 0).sum())

print("Ratios salariales nulos:",
      df_derivadas["Ratio_salario_coste_vida"].isna().sum())
```

La fórmula es:
\[
\text{Ratio salario/coste de vida}=\frac{\text{Salario mensual}}{\text{Índice de coste de vida}}
\]
utilizado  .replace(0, float("nan")) para evitar divisiones por cero.

 Joana tiene un ratio de 22,95 y Carmen de 36,60. Estos valores son indicadores relativos, no importes en euros ni medidas directas del poder adquisitivo.

```python

# PASO 5 - PESO DE RESPONSABILIDAD FAMILIAR
# ==================================================

# 1. Crear la tercera variable derivada
# Multiplicamos el número de hijos por la edad

df_derivadas["Peso_responsabilidad_familiar"] = (
    df_derivadas["Nombre_fills"] * df_derivadas["Edad"]
)

# 2. Mostrar los primeros 10 resultados
display(
    df_derivadas[
        ["Nom", "Nombre_fills", "Edad",
         "Peso_responsabilidad_familiar"]
    ].head(10).style.format({
        "Nombre_fills": "{:.0f}",
        "Edad": "{:.0f}",
        "Peso_responsabilidad_familiar": "{:.0f}"
    }, na_rep="-")
)

# 3. Comprobar casos límite

# Contar edades nulas
print("Edades nulas:",
      df_derivadas["Edad"].isna().sum())

# Detectar edades fuera del rango de revisión
print("Edades fuera del rango 18-100:",
      ((df_derivadas["Edad"] < 18) |
       (df_derivadas["Edad"] > 100)).sum())

# Contar registros sin información sobre hijos
print("Número de hijos nulo:",
      df_derivadas["Nombre_fills"].isna().sum())

# Contar pesos familiares nulos
print("Pesos familiares nulos:",
      df_derivadas["Peso_responsabilidad_familiar"].isna().sum())

# Detectar pesos familiares negativos
print("Pesos familiares negativos:",
      (df_derivadas["Peso_responsabilidad_familiar"] < 0).sum())
```

Se ha creado la tercera variable derivada, Peso_responsabilidad_familiar, multiplicando el número de hijos por la edad del empleado.

Concretament, hauràs de:

» Utilitzant un bucle, exportar en 4 fitxers (format .csv o .xlsx) les dades corresponents a cada Grup Professional. Per exemple: dades_GrupA.csv, dades_GrupB.csv...
» Crear i exportar un cinquè DataFrame (format .csv o .xlsx) que contingui, per a cada Grup Professional:
Nombre total de treballadors.
Salari mitjà.
Edat mediana.

```python

# PASO 1 - COMPROBAR GRUPOS PROFESIONALES
# ==================================================

# Mostrar los grupos profesionales disponibles
print("GRUPOS PROFESIONALES:")
print(df_derivadas["Grup Professional"].value_counts())

# Comprobar las dimensiones del dataset
print("\nDIMENSIONES DEL DATASET:")
print(df_derivadas.shape)

# Comprobar las columnas necesarias
print("\nCOLUMNAS NECESARIAS:")

columnas = ["Grup Professional", "Salari mensual", "Edad"]

for columna in columnas:
    print(columna, ":", columna in df_derivadas.columns)
```

```python
# ==================================================
# SPRINT 8 - NIVEL 2 - EJERCICIO 4
# PASO 2 - EXPORTAR LOS GRUPOS PROFESIONALES
# ==================================================

# Crear una lista con los cuatro grupos
grupos = ["Grup A", "Grup B", "Grup C", "Grup D"]

# Recorrer cada grupo mediante un bucle
for grupo in grupos:

    # Seleccionar los trabajadores del grupo
    df_grupo = df_derivadas[
        df_derivadas["Grup Professional"] == grupo
    ]

    # Crear el nombre del archivo
    nombre_archivo = "datos_" + grupo.replace(" ", "") + ".csv"

    # Exportar los datos a CSV
    df_grupo.to_csv(
        nombre_archivo,
        index=False,
        encoding="utf-8-sig",
        sep=";"
    )

    # Comprobar la exportación
    print(nombre_archivo, "- Trabajadores:", len(df_grupo))
```

```python
# ==================================================
# SPRINT 8 - NIVEL 2 - EJERCICIO 4
# PASO 3 - CREAR TABLA RESUMEN
# ==================================================

# Agrupar por grupo profesional y calcular estadísticas
df_resumen = (
    df_derivadas.groupby("Grup Professional")
    .agg(
        Trabajadores=("DNI", "count"),
        Salario_medio=("Salari mensual", "mean"),
        Edad_mediana=("Edad", "median")
    )
    .reset_index()
)

# Mostrar la tabla resumen
print("RESUMEN POR GRUPO PROFESIONAL:")

display(
    df_resumen.style.format({
        "Trabajadores": "{:.0f}",
        "Salario_medio": "{:,.2f} €",
        "Edad_mediana": "{:.1f}"
    }, thousands=".", decimal=",", na_rep="-")
)
```
