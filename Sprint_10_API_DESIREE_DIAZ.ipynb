PREGUNTA 1 : CONSULTAR POSTS, USERS Y TODOS MEDIANTE GET
PASO 1: IMPORTAR REQUESTS Y DEFINIR LA URL

```python

import requests


url_base = "https://jsonplaceholder.typicode.com"


print("API que vamos a consultar:", url_base)
```

PASO 2: CONSULTAR LAS PUBLICACIONES CON GET

```python

respuesta_posts = requests.get(f"{url_base}/posts")

posts = respuesta_posts.json()

print("PRIMERA PUBLICACIÓN:")
print(posts[0])
```

PASO 3: CONSULTAR USUARIOS Y TAREAS

```python

respuesta_users = requests.get(f"{url_base}/users")
users = respuesta_users.json()
respuesta_todos = requests.get(f"{url_base}/todos")
todos = respuesta_todos.json()
print("PRIMER USUARIO:")
print(users[0])
print("\nPRIMERA TAREA:")
print(todos[0])

```

PASO 4: CONTAR REGISTROS Y COMPROBAR ESTADOS HTTP

```python


print("PUBLICACIONES")
print("Cantidad:", len(posts))
print("Código HTTP:", respuesta_posts.status_code)


print("\nUSUARIOS")
print("Cantidad:", len(users))
print("Código HTTP:", respuesta_users.status_code)


print("\nTAREAS")
print("Cantidad:", len(todos))
print("Código HTTP:", respuesta_todos.status_code)
```

 PASO 5: COMPROBAR EL ERROR HTTP 404

```python


respuesta_error = requests.get(f"{url_base}/posts/999999")


print("Código de estado HTTP:", respuesta_error.status_code)
```

JSONPlaceholder tiene 100 publicaciones de ejemplo, con identificadores del 1 al 100.

Por eso elegimos 999999: sabemos que esa publicación no existe.

PASO 6: REALIZAR UNA PETICIÓN POST

```python

nueva_publicacion = {
    "titulo": "Mi primera API",
    "Cuerpo": "Cuerpo del mensaje_prueba",
    "user_Id": 1
}

respuesta_post = requests.post(
    f"{url_base}/posts",
    json=nueva_publicacion
)

print("RESPUESTA JSON:")
print(respuesta_post.json())

print("\nCódigo de estado HTTP:", respuesta_post.status_code)
```

PASO 7: REALIZAR UNA PETICIÓN PATCH

```python

datos_modificados = {
    "title": "Título actualizado desde Python"
}

respuesta_patch = requests.patch(
    f"{url_base}/posts/1",
    json=datos_modificados
)

print("RESPUESTA JSON:")
print(respuesta_patch.json())


print("\nCódigo de estado HTTP:", respuesta_patch.status_code)
```

PASO 8: REALIZAR UNA PETICIÓN DELETE

```python


respuesta_delete = requests.delete(
    f"{url_base}/posts/1"
)


print("RESPUESTA JSON:")
print(respuesta_delete.json())


print("\nCódigo de estado HTTP:", respuesta_delete.status_code)
```

 SPRINT 10 – NIVEL 2 – EJERCICIO 2
 Documentación de la API Frankfurter v2

**1. Endpoints disponibles**

- `https://api.frankfurter.dev/v2/rates`: permite consultar los tipos de cambio de las divisas.
- `https://api.frankfurter.dev/v2/currencies`: permite consultar las divisas disponibles en la API.

**2. Filtros y parámetros opcionales**

En el endpoint `/v2/rates` podemos utilizar:

- `base`: selecciona la moneda de origen. Ejemplo: `EUR`.
- `quotes`: selecciona la moneda de destino. Ejemplo: `USD`.
- `date`: permite consultar tipos de cambio de una fecha determinada. Ejemplo: `2026-11-06`.

Ejemplo de consulta:

`https://api.frankfurter.dev/v2/rates?base=EUR&quotes=USD`

**3. Formato de respuesta**

La API devuelve los datos en formato **JSON**.

La respuesta contiene campos como `date`, `base`, `quote` y `rate`.

El formato JSON permite interpretar los datos con Python y convertirlos posteriormente en un DataFrame de Pandas.

**Conclusión:** Frankfurter v2 cumple los requisitos del ejercicio porque dispone de varios endpoints, admite parámetros de consulta y devuelve información en formato JSON.

PASO 2: Verifica que la resposta sigui en format JSON (és el més adequat per a aquest exercici)

```python

import requests


url = "https://api.frankfurter.dev/v2/rates"


parametros = {
    "base": "EUR",
    "quotes": "USD"
}


respuesta = requests.get(
    url,
    params=parametros,
    timeout=15
)


print("Código HTTP:", respuesta.status_code)


if respuesta.status_code == 200:


    datos = respuesta.json()


    print("Datos obtenidos:")
    print(datos)

else:
    print("Error al conectar con la API")
```

PASO 3: MOSTRAR LOS CAMPOS DE LA RESPUESTA JSON

```python

print("Código HTTP:", respuesta.status_code)


if respuesta.status_code == 200:


    registro = datos[0]


    print("Fecha:", registro["date"])
    print("Moneda base:", registro["base"])
    print("Moneda destino:", registro["quote"])
    print("Tipo de cambio:", registro["rate"])

else:
    print("Error al consultar la API")
```

PASO 4: CREAR EL DATAFRAME Y MOSTRAR LAS PRIMERAS FILA

```python


import pandas as pd


df_divisas = pd.DataFrame(datos)


display(df_divisas.head())
```

PREGUNTA 1 : ¿CÓMO BUSCAR UN DATASET SOBRE COMPRAVENTAS INMOBILIARIAS?

PASO 1: BUSCAR DATASETS CON PACKAGE_SEARCH

```python

import requests


url_base = "https://opendata-ajuntament.barcelona.cat/data/api/action/"


url_busqueda = url_base + "package_search"


parametros = {"q": "habitatge 2026"}

respuesta = requests.get(
    url_busqueda,
    params=parametros,
    timeout=15
)


print("Código HTTP:", respuesta.status_code)
respuesta.raise_for_status()


datos_busqueda = respuesta.json()


resultados = datos_busqueda["result"]["results"]


for i, dataset in enumerate(resultados[:10]):
    print("Número:", i)
    print("Título:", dataset["title"])
    print("ID:", dataset["id"])
    print("-" * 50)
```

PASO 2: CONSULTAR LOS DETALLES CON PACKAGE_SHOW

```python

numero_dataset = 0


dataset_id = resultados[numero_dataset]["id"]

url_detalle = url_base + "package_show"


respuesta_detalle = requests.get(
    url_detalle,
    params={"id": dataset_id},
    timeout=15
)


print("Código HTTP:", respuesta_detalle.status_code)
respuesta_detalle.raise_for_status()


datos_detalle = respuesta_detalle.json()


dataset_seleccionado = datos_detalle["result"]


print("Título:", dataset_seleccionado["title"])
print("ID:", dataset_seleccionado["id"])
print("Descripción:", dataset_seleccionado.get("notes"))
print("Número de recursos:", len(dataset_seleccionado["resources"]))
```

PREGUNTA 2 : ¿QUÉ RECURSOS CSV O JSON CONTIENE EL DATASET?

PASO 1: IDENTIFICAR LOS RECURSOS DISPONIBLES

```python

recursos = dataset_seleccionado["resources"]


for i, recurso in enumerate(recursos):

    print("Número:", i)
    print("Nombre:", recurso.get("name"))
    print("Formato:", recurso.get("format"))
    print("ID:", recurso.get("id"))
    print("-" * 50)
```

PREGUNTA 2 : ¿QUÉ RECURSO CSV SELECCIONAMOS?

PASO 2: SELECCIONAR UN RECURSO CSV

```python

recurso_seleccionado = recursos[0]


resource_id = recurso_seleccionado["id"]


print("Nombre:", recurso_seleccionado["name"])
print("Formato:", recurso_seleccionado["format"])
print("ID:", resource_id)
```

PREGUNTA 2 : ¿QUÉ INFORMACIÓN CONTIENE EL DATASET?

PASO 3: INSPECCIONAR LAS CLAVES DE PACKAGE_SHOW

```python



print("CLAVES PRINCIPALES:")
print(datos_detalle.keys())


print("\nCLAVES DENTRO DE RESULT:")
print(datos_detalle["result"].keys())


print("\nCLAVES DEL PRIMER RECURSO:")
print(datos_detalle["result"]["resources"][0].keys())
```

PREGUNTA 3 : ¿CÓMO RECUPERAR AL MENOS 100 REGISTROS?

PASO 1: CONSULTAR 100 REGISTROS CON DATASTORE_SEARCH

```python

url_datos = url_base + "datastore_search"


parametros = {
    "resource_id": resource_id,
    "limit": 100
}


respuesta_datos = requests.get(
    url_datos,
    params=parametros,
    timeout=15
)


print("Código HTTP:", respuesta_datos.status_code)
respuesta_datos.raise_for_status()


datos = respuesta_datos.json()


print("Consulta correcta:", datos.get("success"))


if datos.get("success"):
    registros = datos["result"]["records"]
    print("Registros recuperados:", len(registros))
else:
    print("Error:", datos.get("error"))
```

PREGUNTA 4 : ¿CÓMO CONVERTIR LOS RESULTADOS A UN DATAFRAME?

PASO 1: CREAR UN DATAFRAME CON PANDAS

```python

import pandas as pd


df_compraventas = pd.DataFrame(registros)


display(df_compraventas.head())


print("Filas y columnas:", df_compraventas.shape)
```

PREGUNTA 5 : ¿CÓMO GUARDAR EL DATAFRAME EN UN ARCHIVO CSV?

 PASO 1: EXPORTAR EL DATAFRAME A CSV

```python

df_compraventas.to_csv(
    "compraventas_barcelona_2025.csv",
    index=False,
    encoding="utf-8-sig"
)


print("Archivo CSV guardado correctamente.")
print("Registros exportados:", len(df_compraventas))
```
