```python
# ==========================================================
# SPRINT 9 - CONECTAR CON MYSQL
# ==========================================================

import pandas as pd
import matplotlib.pyplot as plt
import mysql.connector
from getpass import getpass

print("Librerías importadas correctamente")

# Conexión a MySQL
conexion = mysql.connector.connect(
    host="localhost",
    port=3306,
    user="root",
    password=getpass("Contraseña MySQL: "),
    database="transactions"
)

# Comprobar conexión
print("Conexión correcta:", conexion.is_connected())
```

```python
# ==========================================================
# SPRINT 9 - NIVEL 1 - EJERCICIO 1
# PASO 2. CARGAR DATOS DESDE MYSQL A PANDAS
# ==========================================================

# Cargar la tabla de transacciones
df_transacciones = pd.read_sql(
    "SELECT * FROM transactions",
    conexion
)

# Cargar la tabla de compañías
df_companies = pd.read_sql(
    "SELECT * FROM companies",
    conexion
)

# Mostrar las primeras cinco filas
print("TABLA TRANSACTIONS")
display(df_transacciones.head())

print("TABLA COMPANIES")
display(df_companies.head())

# Comprobar las dimensiones
print("Dimensiones transacciones:", df_transacciones.shape)
print("Dimensiones compañías:", df_companies.shape)
```

```python
# PASO 8. COMPROBAR LOS TIPOS DE DATOS

df_transacciones.info()
```

```python
# ==================================================
# SPRINT 9 - NIVEL 1 - EJERCICIO 1
# PREGUNTA 1: EVOLUCIÓN DE LAS TRANSACCIONES
# ==================================================

# PASO 1. COMPROBAR LA COLUMNA DE FECHAS

print(df_transacciones["timestamp"].dtype)

display(df_transacciones[["timestamp"]].head())
```

```python
# PASO 2. CONVERTIR LA COLUMNA TIMESTAMP A FECHA

df_transacciones["timestamp"] = pd.to_datetime(
    df_transacciones["timestamp"],
    errors="coerce"
)

# Comprobar el nuevo tipo de dato
print(df_transacciones["timestamp"].dtype)

# Mostrar las primeras cinco fechas
display(df_transacciones[["timestamp"]].head())
```

```python
# PASO 3. COMPROBAR FECHAS NULAS

# Contar las fechas nulas
fechas_nulas = df_transacciones["timestamp"].isna().sum()

# Mostrar el resultado
print("Número de fechas nulas:", fechas_nulas)
```

```python
# PASO 4. AGRUPAR LAS TRANSACCIONES POR MES

# Agrupar las transacciones por mes y contarlas
transacciones_mes = (
    df_transacciones
    .dropna(subset=["timestamp"])
    .groupby(df_transacciones["timestamp"].dt.to_period("M"))
    .size()
)

# Mostrar los resultados
display(transacciones_mes)
```

```python
# PASO 5. GRÁFICO DE EVOLUCIÓN DE TRANSACCIONES

# Importar la librería de gráficos
import matplotlib.pyplot as plt

# Crear el gráfico de líneas
transacciones_mes.plot(
    kind="line",
    figsize=(12, 5),
    marker="o"
)

# Añadir título y nombres de los ejes
plt.title("Evolución mensual de las transacciones")
plt.xlabel("Mes")
plt.ylabel("Número de transacciones")

# Mostrar cuadrícula
plt.grid(True)

# Mostrar el gráfico
plt.show()
```

Utilizamos un gráfico de líneas porque permite analizar la evolución de una variable numérica a lo largo del tiempo.

En nuestro caso, estudiamos el número de transacciones agrupadas por mes.

```python
# ==========================================================
# SPRINT 9 - NIVEL 1 - EJERCICIO 2
# PASO 1. VISUALIZAR LAS DOS TABLAS
# ==========================================================

# Mostrar las primeras 5 filas de transacciones
print("TABLA TRANSACTIONS")
display(df_transacciones.head())

# Mostrar las primeras 5 filas de compañías
print("TABLA COMPANIES")
display(df_companies.head())

# Mostrar las columnas de transacciones
print("COLUMNAS TRANSACTIONS:")
print(df_transacciones.columns.tolist())

# Mostrar las columnas de compañías
print("\nCOLUMNAS COMPANIES:")
print(df_companies.columns.tolist())
```

```python
# ==========================================================
# SPRINT 9 - NIVEL 1 - EJERCICIO 2
# PREGUNTA 2 - PASO 1
# COMBINAR TRANSACCIONES Y COMPAÑÍAS
# ==========================================================

# Combinar las dos tablas
df_ventas_paises = pd.merge(
    df_transacciones,
    df_companies[["company_id", "company_name", "country"]],
    left_on="business_id",
    right_on="company_id",
    how="left"
)

# Mostrar las primeras cinco filas
display(
    df_ventas_paises[
        ["business_id", "company_name", "country", "amount"]
    ].head()
)

# Comprobar dimensiones
print("Dimensiones:", df_ventas_paises.shape)

# Comprobar transacciones sin país
print(
    "Transacciones sin país:",
    df_ventas_paises["country"].isna().sum()
)
```

```python
# ==========================================================
# SPRINT 9 - NIVEL 1 - EJERCICIO 2
# PREGUNTA 1 - PASO 2
# COMPROBAR LA UNIÓN DE LAS TABLAS
# ==========================================================

# Número de transacciones originales
print("Transacciones originales:", len(df_transacciones))

# Número de filas después del merge
print("Transacciones después del merge:", len(df_ventas_paises))

# Comprobar si se han añadido filas
print(
    "Diferencia de filas:",
    len(df_ventas_paises) - len(df_transacciones)
)
```

```python
# ==========================================================
# SPRINT 9 - NIVEL 1 - EJERCICIO 2
# PREGUNTA 1 - PASO 3
# COMPROBAR LOS DATOS DE FACTURACIÓN
# ==========================================================

# Comprobar el tipo de dato del importe :Con esto comprobaremos si amount permite hacer sumas y cuántas transacciones están marcadas como rechazadas.
print("Tipo de dato de amount:")
print(df_ventas_paises["amount"].dtype)

# Comprobar los valores de declined : groupby("country") y sum() para calcular la facturación por país, excluyendo las operaciones rechazadas si confirmamos que declined = 1 significa rechazo.
print("\nValores de declined:")
print(df_ventas_paises["declined"].value_counts())


#Vamos a considerar que declined = 1 representa una transacción rechazada, por lo que no la incluiremos en la facturación.
```

```python
# ==========================================================
# SPRINT 9 - NIVEL 1 - EJERCICIO 2
# PREGUNTA 1 - PASO 4
# CALCULAR FACTURACIÓN POR PAÍS
# ==========================================================

# Seleccionar las transacciones no rechazadas
df_ventas_validas = df_ventas_paises[
    df_ventas_paises["declined"] == 0
].copy()

# Agrupar por país y sumar los importes
facturacion_pais = (
    df_ventas_validas
    .groupby("country")["amount"]
    .sum()
    .reset_index()
)

# Ordenar de mayor a menor facturación
facturacion_pais = facturacion_pais.sort_values(
    by="amount",
    ascending=False
)

# Mostrar resultados
display(facturacion_pais)

# VERSIÓN 1. Mostrar los países automáticamente desde la tabla
print(
    "Los países con mayor facturación son: "
    + ", ".join(facturacion_pais["country"].head(5))
)

# VERSIÓN 2. Mostrar la conclusión escrita en español
#print(
    #"Los países con mayor facturación son Suecia, "
    #"Países Bajos, Reino Unido, Italia y Alemania."
#)
```

```python
# ==========================================================
# SPRINT 9 - NIVEL 1 - EJERCICIO 2
# PREGUNTA 1 - PASO 5
# CALCULAR PORCENTAJES DE FACTURACIÓN
# ==========================================================

# Calcular la facturación total
total_facturacion = facturacion_pais["amount"].sum()

# Calcular el porcentaje de cada país
facturacion_pais["porcentaje"] = (
    facturacion_pais["amount"] / total_facturacion * 100
)

# Calcular el porcentaje acumulado: Este cálculo nos permitirá analizar si unos pocos países concentran la mayor parte de los ingresos.
facturacion_pais["porcentaje_acumulado"] = (
    facturacion_pais["porcentaje"].cumsum()
)

# Redondear los porcentajes a dos decimales
facturacion_pais["porcentaje"] = facturacion_pais["porcentaje"].round(2)

facturacion_pais["porcentaje_acumulado"] = (
    facturacion_pais["porcentaje_acumulado"].round(2)
)

# Mostrar los resultados
display(facturacion_pais)


```

```python
# ==========================================================
# SPRINT 9 - NIVEL 1 - EJERCICIO 2
# PREGUNTA 3 : ¿Hay usuarios que han dejado de comprar recientemente?
# PASO 2: CALCULAR LOS DÍAS DESDE LA ÚLTIMA COMPRA
# ==========================================================
#Para calcular los días transcurridos, utilizaremos como referencia la fecha más reciente del dataset, ya que nuestras transacciones son históricas.


# Obtener la fecha más reciente del dataset
fecha_referencia = df_compras_validas["timestamp"].max()

# Calcular los días desde la última compra de cada usuario
ultima_compra["dias_sin_comprar"] = (
    fecha_referencia - ultima_compra["ultima_compra"]
).dt.days

# Mostrar la fecha de referencia
print("FECHA DE REFERENCIA:")
print(fecha_referencia)

# Mostrar los primeros diez usuarios
print("\nDÍAS DESDE LA ÚLTIMA COMPRA:")
display(ultima_compra.head(10))
```

```python
# con la tabla anterior: podemos identificar productos y analizar la captación de usuarios, pero no calcular los ingresos reales de cada producto sin sus precios individuales o el desglose de los importes.
#Hay que separar productos

# ==========================================================
# SPRINT 9 - NIVEL 2 - EJERCICIO 1
# PREGUNTA 1 : ¿Los productos que generan más ingresos también captan nuevos usuarios?
# PASO 3: SEPARAR LOS PRODUCTOS DE CADA TRANSACCIÓN
# ==========================================================

# Seleccionar las transacciones no rechazadas
df_productos = df_transacciones[
    df_transacciones["declined"] == 0
].copy()

# Separar los identificadores de productos
df_productos["product_ids"] = (
    df_productos["product_ids"]
    .str.split(",")
)

# Crear una fila por cada producto
df_productos = df_productos.explode("product_ids")

# Eliminar espacios y convertir el identificador a número
df_productos["product_ids"] = pd.to_numeric(
    df_productos["product_ids"].str.strip(),
    errors="coerce"
)

# Mostrar las primeras diez filas
print("TRANSACCIONES SEPARADAS POR PRODUCTO:")

display(
    df_productos[
        ["id", "user_id", "product_ids", "amount", "timestamp"]
    ].head(10)
)
```

**Paso 4. Identificar los nuevos usuarios**
El enunciado habla de la primera transacción en el sistema, no necesariamente de la primera compra aceptada. Por tanto, para identificar correctamente a los usuarios nuevos, debemos buscar primero su transacción más antigua en df_transacciones y después relacionarla con los productos.

Debemos identificar primero la fecha de su primera compra y después comprobar qué productos compró en esa transacción.

identificar la primera transacción cronológica del usuario en todo el sistema.

```python
# ==========================================================
# SPRINT 9 - NIVEL 2 - EJERCICIO 1
# PREGUNTA 1 : ¿Los productos que generan más ingresos también captan nuevos usuarios?
# PASO 4: IDENTIFICAR LOS NUEVOS USUARIOS
# ==========================================================

# Crear una copia de todas las transacciones
df_primera_transaccion = df_transacciones.copy()

# Convertir las fechas a datetime
df_primera_transaccion["timestamp"] = pd.to_datetime(
    df_primera_transaccion["timestamp"]
)

# Ordenar las transacciones cronológicamente
df_primera_transaccion = df_primera_transaccion.sort_values(
    by=["timestamp", "id"]
)

# Identificar la primera transacción de cada usuario
primeras_transacciones = (
    df_primera_transaccion
    .drop_duplicates(subset="user_id", keep="first")
)

# Obtener los identificadores de esas transacciones
ids_primera_transaccion = primeras_transacciones["id"]

# Marcar si cada transacción corresponde a un nuevo usuario
df_productos["es_nou_usuari"] = (
    df_productos["id"].isin(ids_primera_transaccion)
)

# Mostrar los resultados
print("PRODUCTOS Y CAPTACIÓN DE NUEVOS USUARIOS:")

display(
    df_productos[
        ["user_id", "product_ids", "timestamp", "es_nou_usuari"]
    ].head(10)
)

# Comprobar los valores True y False
print("\nVALORES DE es_nou_usuari:")
print(df_productos["es_nou_usuari"].value_counts())
```

**Paso 5. Calcular la captación de usuarios por producto**

1. Número de usuarios nuevos que ha captado cada producto.
2. Porcentaje de usuarios nuevos respecto al total de compradores de ese producto.

```python
# ==========================================================
# SPRINT 9 - NIVEL 2 - EJERCICIO 1
# PREGUNTA 1 : ¿Los productos que generan más ingresos también captan nuevos usuarios?
# PASO 5: CALCULAR LA CAPTACIÓN DE USUARIOS POR PRODUCTO
# ==========================================================

# Contar los compradores únicos de cada producto
compradores_producto = (
    df_productos
    .groupby("product_ids")["user_id"]
    .nunique()
    .reset_index(name="total_compradores")
)

# Contar los usuarios nuevos de cada producto
nuevos_producto = (
    df_productos[df_productos["es_nou_usuari"] == True]
    .groupby("product_ids")["user_id"]
    .nunique()
    .reset_index(name="nuevos_usuarios")
)

# Combinar las dos tablas
captacion_productos = pd.merge(
    compradores_producto,
    nuevos_producto,
    on="product_ids",
    how="left"
)

# Sustituir por cero los productos sin usuarios nuevos
captacion_productos["nuevos_usuarios"] = (
    captacion_productos["nuevos_usuarios"].fillna(0).astype(int)
)

# Calcular el porcentaje de captación
captacion_productos["porcentaje_nuevos"] = (
    captacion_productos["nuevos_usuarios"]
    / captacion_productos["total_compradores"]
    * 100
).round(2)

# Ordenar por número de nuevos usuarios
captacion_productos = captacion_productos.sort_values(
    by="nuevos_usuarios",
    ascending=False
)

# Mostrar los resultados
print("CAPTACIÓN DE NUEVOS USUARIOS POR PRODUCTO:")
display(captacion_productos.head(15))
```

Observamos que los porcentajes de captación son relativamente bajos

**PASO 6: CALCULAR EL IMPORTE ASOCIADO Y LA VARIABILIDAD**

Como no tenemos los precios individuales de los productos, estamos utilizando el importe total de las transacciones en las que aparece cada producto.

Queremos conocer qué productos aparecen en las compras de mayor importe, cuánto varían esos importes y cuántas transacciones incluyen cada producto.

```python
# ==========================================================
# SPRINT 9 - NIVEL 2 - EJERCICIO 1
# PREGUNTA 1 : ¿Los productos que generan más ingresos también captan nuevos usuarios?
# PASO 6: CALCULAR EL IMPORTE ASOCIADO Y LA VARIABILIDAD
# ==========================================================

# Agrupar por producto y calcular las métricas
metricas_productos = (
    df_productos
    .groupby("product_ids")
    .agg(
        importe_asociado=("amount", "sum"),
        variabilidad_importe=("amount", "std"),
        num_transacciones=("id", "nunique")
    )
    .reset_index()
)

# Combinar las métricas con la captación de usuarios
analisis_productos = pd.merge(
    metricas_productos,
    captacion_productos,
    on="product_ids",
    how="left"
)

# Ordenar por importe asociado, de mayor a menor
analisis_productos = analisis_productos.sort_values(
    by="importe_asociado",
    ascending=False
)

# Ordenar por importe asociado, de mayor a menor
analisis_productos = analisis_productos.sort_values(
    by="importe_asociado",
    ascending=False
)

# Mostrar los 15 primeros productos sin índice
print(analisis_productos.head(15).to_string(index=False))
```

El importe_asociado no representa los ingresos individuales reales del producto, porque una misma transacción puede contener varios productos.


**Conclusión pregunta 1 – Ejercicio 1 – Nivel 2 (Paso 6):**

Los resultados muestran que el producto 67 presenta un importe asociado de 1.141.422,71 €, una variabilidad de 149,66 € y un total de 2.525 transacciones. Este producto ha sido adquirido por 1.962 compradores únicos, de los cuales 118 son usuarios nuevos, lo que representa un 6,01 % del total de compradores.

En conclusión, el producto 67 aparece en transacciones de elevado importe, pero su porcentaje de captación de nuevos usuarios es relativamente reducido. Esto indica que la mayoría de sus compradores son usuarios que ya habían realizado alguna transacción anteriormente.

Es importante señalar que el importe asociado corresponde al valor total de las transacciones en las que aparece el producto y no representa su facturación individual real, ya que una misma transacción puede incluir varios productos.

```python
#Ahora que tenemos calculados los importes asociados y la captación de usuarios, vamos a clasificar los productos en cuatro categorías.

# ==========================================================
# SPRINT 9 - NIVEL 2 - EJERCICIO 1
# PREGUNTA 1 : ¿Los productos que generan más ingresos también captan nuevos usuarios?
# PASO 7: CLASIFICAR LOS PRODUCTOS
# ==========================================================

# 1. Calcular las medianas
mediana_importe = analisis_productos["importe_asociado"].median()

mediana_captacion = analisis_productos["porcentaje_nuevos"].median()


# 2. Crear una función para clasificar los productos
def clasificar_producto(fila):

    if fila["importe_asociado"] >= mediana_importe:

        if fila["porcentaje_nuevos"] >= mediana_captacion:
            return "Captador"

        else:
            return "Recurrente"

    else:

        if fila["porcentaje_nuevos"] >= mediana_captacion:
            return "Emergente"

        else:
            return "Estancado"


# 3. Aplicar la función a cada producto
analisis_productos["categoria"] = analisis_productos.apply(
    clasificar_producto,
    axis=1
)


# 4. Mostrar los primeros 15 productos sin índice
print("CLASIFICACIÓN DE PRODUCTOS:")

print(analisis_productos.head(15).to_string(index=False))


# 5. Contar los productos por categoría
print("\nNÚMERO DE PRODUCTOS POR CATEGORÍA:")

print(analisis_productos["categoria"].value_counts().to_string())
```

Los productos recurrentes y emergentes son los grupos más numerosos. Esto indica que no todos los productos asociados a compras de mayor importe presentan también una captación elevada de nuevos usuarios.
