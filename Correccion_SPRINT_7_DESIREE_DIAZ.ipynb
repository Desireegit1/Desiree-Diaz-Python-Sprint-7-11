```python


def validar_datos(candidato):


    campos = ["experiencia", "catalan", "excel", "titulacion"]

    for campo in campos:
        if not candidato[campo].isdigit():
            print("Error: el campo", campo, "debe ser un número entero.")
            return None


    for campo in campos:
        candidato[campo] = int(candidato[campo])


    if candidato["experiencia"] < 0:
        print("Error: la experiencia no puede ser negativa.")
        return None


    for campo in ["catalan", "excel", "titulacion"]:
        if not 0 <= candidato[campo] <= 3:
            print("Error:", campo, "debe estar entre 0 y 3.")
            return None

    return candidato
```

```python
def validar_datos(candidato):

    match candidato:

        case {"nombre": ""}:
            return False

        case {"telefono": ""}:
            return False

        case {"experiencia": experiencia} if experiencia < 0:
            return False

        case {"catalan": catalan} if catalan < 0 or catalan > 3:
            return False

        case {"excel": excel} if excel < 0 or excel > 3:
            return False

        case {"titulacion": titulacion} if titulacion < 0 or titulacion > 3:
            return False

        case _:
            return True
```

```python
def calcular_puntuacion(candidato):

    puntuacion = (
        candidato["experiencia"] * 1.5
        + candidato["catalan"]
        + candidato["excel"]
        + candidato["titulacion"]
    )

    return puntuacion
```

```python
def clasificar_candidato(puntuacion):

    if puntuacion >= 12:
        return "Prioridad Alta"

    elif puntuacion >= 8:
        return "Candidato adecuado"

    else:
        return "No prioritario"
```

```python
def mostrar_informe(candidatos):

    prioridades = {
        "Prioridad Alta": 3,
        "Candidato adecuado": 2,
        "No prioritario": 1
    }

    candidatos_ordenados = sorted(
        candidatos,
        key=lambda candidato: (
            prioridades[candidato["clasificacion"]],
            candidato["puntuacion"]
        ),
        reverse=True
    )

    df = pd.DataFrame(candidatos_ordenados)



    print("----- INFORME FINAL -----")
    display(df)
```

```python
def main_seleccion():

    candidatos = []

    while len(candidatos) < 10:

        candidato = obtener_datos()

        if validar_datos(candidato):

            puntuacion = calcular_puntuacion(candidato)
            clasificacion = clasificar_candidato(puntuacion)

            candidato["puntuacion"] = puntuacion
            candidato["clasificacion"] = clasificacion

            candidatos.append(candidato)

        else:
            print("Datos no válidos. El candidato no se registra.")

    mostrar_informe(candidatos)
```

Creamos un diccionario llamado comportamientos que relaciona los códigos del 0 al 4 con un segmento de cliente, un mensaje personalizado y un factor numérico.

```python
factores_edad = {
    "18-25": 1.20,
    "26-40": 1.00,
    "41-60": 0.85,
    "60+": 0.70
}
```

Creamos un diccionario llamado factores_edad que asigna un factor numérico a cada grupo de edad

```python
def obtener_segmento_y_factor(codigo):

    return comportamientos[codigo]
```

```python
def obtener_factor_edad(grupo_edad):

    return factores_edad[grupo_edad]
```

```python
def calcular_riesgo(factor_codigo, factor_edad):

    return factor_codigo * factor_edad
```

Las tres funciones permiten obtener los factores de comportamiento y edad de un cliente y calcular su riesgo final mediante una multiplicación. De esta manera, organizamos el programa en funciones independientes y facilitamos su mantenimiento y reutilización.

```python
def procesar_clientes(clientes):

    clientes_procesados = []

    for cliente in clientes:

        datos_comportamiento = obtener_segmento_y_factor(
            cliente["codigo_comportamiento"]
        )

        cliente_procesado = {
            "id_cliente": cliente["id_cliente"],
            "nombre": cliente["nombre"],
            "edad": cliente["edad"],
            "grupo_edad": cliente["grupo_edad"],
            "codigo_comportamiento": cliente["codigo_comportamiento"],
            "segmento": datos_comportamiento["segmento"],
            "mensaje": datos_comportamiento["mensaje"],
            "factor_codigo": datos_comportamiento["factor_codigo"],

            "factor_edad": obtener_factor_edad(
                cliente["grupo_edad"]
            ),

            "riesgo_final": calcular_riesgo(
                datos_comportamiento["factor_codigo"],
                obtener_factor_edad(cliente["grupo_edad"])
            )
        }

        clientes_procesados.append(cliente_procesado)

    return clientes_procesados
```

**PREGUNTA** :¿Qué diferencias existen entre clientes con el mismo comportamiento, pero de edades diferentes? Por ejemplo, ¿un cliente joven que abandona la cesta tiene el mismo riesgo que uno de mayor edad?

agrupar los clientes por código de comportamiento y comparar sus factores de edad y sus riesgos finales.

```python


print("\n2. Diferencias entre clientes con el mismo comportamiento:")

comparacion = df.groupby(
    ["codigo_comportamiento", "grupo_edad"]
).agg(
    numero_clientes=("id_cliente", "count"),
    factor_edad=("factor_edad", "first"),
    riesgo_medio=("riesgo_final", "mean")
).reset_index()

display(comparacion)
```

df.groupby(["codigo_comportamiento", "grupo_edad"]) agrupa los clientes que tienen el mismo comportamiento y pertenecen al mismo grupo de edad. utilizamos groupby() Porque necesitamos comparar clientes que comparten el mismo comportamiento, pero pertenecen a grupos de edad diferentes.

.agg() permite realizar cálculos sobre los grupos creados.

numero_clientes=("id_cliente", "count") cuenta cuántos clientes existen en cada grupo.

factor_edad=("factor_edad", "first") obtiene el factor de edad correspondiente al grupo.

riesgo_medio=("riesgo_final", "mean") calcula el riesgo medio de cada combinación de comportamiento y edad.

.reset_index() convierte las columnas utilizadas para agrupar en columnas normales del DataFrame.

display(comparacion) muestra los resultados en una tabla.

```python
def mostrar_respuestas(df):

    print("2. Diferencias entre clientes con el mismo comportamiento:")

    comparacion = df.groupby(
        ["codigo_comportamiento", "grupo_edad"]
    ).agg(
        numero_clientes=("id_cliente", "count"),
        factor_edad=("factor_edad", "first"),
        riesgo_medio=("riesgo_final", "mean")
    ).reset_index()

    display(comparacion)
```

Exercici 1: Generador de contrasenyes segures
Ets analista de dades dins el departament de Ciberseguretat d'una empresa tecnològica. S'ha detectat que la majoria de treballadors continuen utilitzant contrasenyes molt dèbils com “Asdf1234”,
dates d'aniversari o combinacions fàcils d'endevinar. Per aquest motiu, el responsable del departament t'ha demanat desenvolupar un generador de contrasenyes segures, totalment parametritzable i modular.

```python
import random
import string
```

```python
def crear_conjunto_caracteres(
    mayusculas,
    minusculas,
    numeros,
    signos
):

    conjuntos = []

    if mayusculas:
        conjuntos.append(string.ascii_uppercase)

    if minusculas:
        conjuntos.append(string.ascii_lowercase)

    if numeros:
        conjuntos.append(string.digits)

    if signos:
        conjuntos.append(string.punctuation)

    return conjuntos
```

```python
def asegurar_caracteres_obligatorios(conjuntos):

    caracteres_obligatorios = []

    for conjunto in conjuntos:

        caracter = random.choice(conjunto)

        caracteres_obligatorios.append(caracter)

    return caracteres_obligatorios
```

```python
def completar_contrasena(
    caracteres_obligatorios,
    conjuntos,
    longitud
):

    contrasena = caracteres_obligatorios.copy()

    todos_caracteres = "".join(conjuntos)

    while len(contrasena) < longitud:

        caracter = random.choice(todos_caracteres)

        contrasena.append(caracter)

    # Mezclar los caracteres de la contraseña
    random.shuffle(contrasena)

    # Convertir la lista en una cadena de texto
    return "".join(contrasena)
```

```python
def crear_contrasena(
    longitud,
    mayusculas=True,
    minusculas=True,
    numeros=True,
    signos=False
):

    conjuntos = crear_conjunto_caracteres(
        mayusculas,
        minusculas,
        numeros,
        signos
    )


    if not conjuntos:
        return "Error: debes seleccionar al menos un tipo de carácter"

    if not validar_longitud(longitud, conjuntos):
        return "Error: longitud insuficiente"

    caracteres_obligatorios = asegurar_caracteres_obligatorios(
        conjuntos
    )

    contrasena = completar_contrasena(
        caracteres_obligatorios,
        conjuntos,
        longitud
    )

    return contrasena
```

```python
def validar_longitud(longitud, conjuntos):

    cantidad_tipos = len(conjuntos)

    if longitud < cantidad_tipos:
        return False

    return True
```

```python
contrasena = crear_contrasena(12)

print("Contraseña generada:", contrasena)
```

**Exercici 2**:

Processament automàtic de dades esportives
Ara formes part d'un equip d'analítica dins una organització esportiva que gestiona dades de la lliga catalana.
La teva companya de feina vol automatitzar la lectura del seu arxiu històric de resultats i extreure les estadístiques més importants.
T'ha demanat fer un petit sistema en Python que processi el fitxer 📄 "historic partits.txt"
🎯 Objectius
Mitjançant un codi modular, clar i documentat, el teu programa ha de calcular:

» 1.Quin equip és el més golejador.

» 2.Quin és el més golejat.

» 3.El nombre total de gols que ha fet cada equip.

» 4.La classificació final ordenada per punts on cada victòria suma 3 pts, cada empat 1 pts y cada derrota 0 pts.

```python
import pandas as pd
```

```python
def leer_archivo(nombre_archivo):

    with open(nombre_archivo, "r", encoding="utf-8") as archivo:

        lineas = archivo.readlines()

    return lineas
```

```python
def procesar_linea(linea):

    datos = linea.strip().split("\t")

    equipo_local = datos[0]
    resultado = datos[1]
    equipo_visitante = datos[2]

    goles = resultado.split("-")

    goles_local = int(goles[0])
    goles_visitante = int(goles[1])

    return equipo_local, goles_local, goles_visitante, equipo_visitante
```

```python
def inicializar_equipo(estadisticas, equipo):

    if equipo not in estadisticas:

          estadisticas[equipo] = {
            "goles_favor": 0,
            "goles_contra": 0,
            "guanyats": 0,
            "empatats": 0,
            "perduts": 0,
            "puntos": 0
        }
```

```python
def actualizar_estadisticas(
    estadisticas,
    equipo_local,
    goles_local,
    goles_visitante,
    equipo_visitante
):

    # Actualizar goles
    estadisticas[equipo_local]["goles_favor"] += goles_local
    estadisticas[equipo_local]["goles_contra"] += goles_visitante

    estadisticas[equipo_visitante]["goles_favor"] += goles_visitante
    estadisticas[equipo_visitante]["goles_contra"] += goles_local

    # Victoria del equipo local
    if goles_local > goles_visitante:

        estadisticas[equipo_local]["puntos"] += 3
        estadisticas[equipo_local]["guanyats"] += 1
        estadisticas[equipo_visitante]["perduts"] += 1

    # Victoria del equipo visitante
    elif goles_local < goles_visitante:

        estadisticas[equipo_visitante]["puntos"] += 3
        estadisticas[equipo_visitante]["guanyats"] += 1
        estadisticas[equipo_local]["perduts"] += 1

    # Empate
    else:

        estadisticas[equipo_local]["puntos"] += 1
        estadisticas[equipo_visitante]["puntos"] += 1

        estadisticas[equipo_local]["empatats"] += 1
        estadisticas[equipo_visitante]["empatats"] += 1
```

```python
def procesar_partidos(lineas):

    estadisticas = {}

    for linea in lineas:

        equipo_local, goles_local, goles_visitante, equipo_visitante = procesar_linea(
            linea
        )

        inicializar_equipo(
            estadisticas,
            equipo_local
        )

        inicializar_equipo(
            estadisticas,
            equipo_visitante
        )

        actualizar_estadisticas(
            estadisticas,
            equipo_local,
            goles_local,
            goles_visitante,
            equipo_visitante
        )

    return estadisticas
```

```python
def crear_dataframe(estadisticas):

    df = pd.DataFrame.from_dict(
        estadisticas,
        orient="index"
    )

    df = df.reset_index()

    df = df.rename(
        columns={"index": "equipo"}
    )

    return df
```

```python
def ordenar_clasificacion(df):

    df_ordenado = df.sort_values(
        by="puntos",
        ascending=False
    )

    df_ordenado = df_ordenado.reset_index(
        drop=True
    )

    return df_ordenado
```

```python
def obtener_destacados(df):

    equipo_mas_goleador = df.loc[
        df["goles_favor"].idxmax(),
        "equipo"
    ]

    equipo_mas_goleado = df.loc[
        df["goles_contra"].idxmax(),
        "equipo"
    ]

    return equipo_mas_goleador, equipo_mas_goleado
```

```python
def mostrar_resultados(
    df_clasificacion,
    equipo_mas_goleador,
    equipo_mas_goleado
):

    print("CLASIFICACIÓN FINAL:")
    display(df_clasificacion.set_index("equipo"))

    print(
        "\nEquipo más goleador:",
        equipo_mas_goleador
    )

    print(
        "Equipo más goleado:",
        equipo_mas_goleado
    )
```

```python
def main_deportes(ruta_archivo):

    lineas = leer_archivo(ruta_archivo)

    estadisticas = procesar_partidos(lineas)

    df = crear_dataframe(estadisticas)

    df_clasificacion = ordenar_clasificacion(df)

    equipo_mas_goleador, equipo_mas_goleado = obtener_destacados(
        df_clasificacion
    )

    mostrar_resultados(
        df_clasificacion,
        equipo_mas_goleador,
        equipo_mas_goleado
    )
```
