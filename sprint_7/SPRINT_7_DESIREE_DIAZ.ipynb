```python
def obtener_datos():
    nombre = input("Nombre y apellidos: ")
    telefono = input("Teléfono de contacto: ")
    experiencia = int(input("Años de experiencia dentro de la empresa: "))
    catalan = int(input("Conocimiento de catalán (0-3): "))
    excel = int(input("Dominio de Excel (0-3): "))
    titulacion = int(input("Título académico (0-3): "))

    candidato = {
        "nombre": nombre,
        "telefono": telefono,
        "experiencia": experiencia,
        "catalan": catalan,
        "excel": excel,
        "titulacion": titulacion
    }

    return candidato
```

```python
def validar_datos(candidato):

    if candidato["nombre"] == "":
        return False

    if candidato["telefono"] == "":
        return False

    if candidato["experiencia"] < 0:
        return False

    if candidato["catalan"] < 0 or candidato["catalan"] > 3:
        return False

    if candidato["excel"] < 0 or candidato["excel"] > 3:
        return False

    if candidato["titulacion"] < 0 or candidato["titulacion"] > 3:
        return False

    return True
```

```python
def calcular_puntuacion(candidato):

    puntuacion = (
        candidato["experiencia"] * 1.5
        + candidato["catalan"]
        + candidato["excel"]
        + candidato["titulacion"]
    )

    return puntuacion
```

```python
def mostrar_individual(candidato, puntuacion, clasificacion):

    print("Candidato:", candidato["nombre"])
    print("Puntuación:", puntuacion)
    print("Clasificación:", clasificacion)
    print("-----------------------------")
```

```python
def clasificar_candidato(puntuacion):

    if puntuacion >= 12:
        return "Prioridad Alta"

    elif puntuacion >= 8:
        return "Candidato adecuado"

    else:
        return "No prioritario"
```

```python
def mostrar_informe(candidatos):

    prioridades = {
        "Prioridad Alta": 3,
        "Candidato adecuado": 2,
        "No prioritario": 1
    }

    candidatos_ordenados = sorted(
        candidatos,
        key=lambda candidato: (
            prioridades[candidato["clasificacion"]],
            candidato["puntuacion"]
        ),
        reverse=True
    )

    print("----- INFORME FINAL -----")

    for candidato in candidatos_ordenados:
        print("Nombre:", candidato["nombre"])
        print("Teléfono:", candidato["telefono"])
        print("Puntuación:", candidato["puntuacion"])
        print("-------------------------")
```

```python
def main_seleccion():

    candidatos = []

    while len(candidatos) < 10:

        candidato = obtener_datos()

        if validar_datos(candidato):

            puntuacion = calcular_puntuacion(candidato)
            clasificacion = clasificar_candidato(puntuacion)

            candidato["puntuacion"] = puntuacion
            candidato["clasificacion"] = clasificacion

            candidatos.append(candidato)

            mostrar_individual(
                candidato,
                puntuacion,
                clasificacion
            )

        else:
            print("Datos no válidos. El candidato no se registra.")

    mostrar_informe(candidatos)
```

```python
# -*- coding: utf-8 -*-

import pandas as pd

pacientes = [
    "Maria|45|tos, Fiebre Alta, dolor_pecho",
    "Luis|33|dolor_cabeza,faTiga",
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos",
    "Jordi|52|fatiga , tos",
    "Anna|29|fiebre alta,dolor_cabeza"
]

niveles_riesgo = {
    "fiebre_alta": 3,
    "dificultad_respirar": 5,
    "dolor_pecho": 4,
    "tos": 1,
    "fatiga": 1,
    "dolor_cabeza": 1
}
```

```python
def normalizar_sintomas(texto_sintomas):

    sintomas = texto_sintomas.split(",")

    sintomas_normalizados = []

    for sintoma in sintomas:

        sintoma = sintoma.strip()
        sintoma = sintoma.lower()
        sintoma = sintoma.replace(" ", "_")

        sintomas_normalizados.append(sintoma)

    return sintomas_normalizados
```

```python
def convertir_paciente(registro):

    datos = registro.split("|")

    nombre = datos[0]
    edad = int(datos[1])
    sintomas = normalizar_sintomas(datos[2])

    paciente = {
        "nombre": nombre,
        "edad": edad,
        "sintomas": sintomas
    }

    return paciente
```

```python
def calcular_riesgo(sintomas):

    riesgo_total = 0

    for sintoma in sintomas:

        riesgo_total = riesgo_total + niveles_riesgo[sintoma]

    return riesgo_total
```

```python
def procesar_pacientes(pacientes):

    pacientes_procesados = []

    for registro in pacientes:

        paciente = convertir_paciente(registro)

        riesgo_total = calcular_riesgo(
            paciente["sintomas"]
        )

        paciente["riesgo_total"] = riesgo_total

        pacientes_procesados.append(paciente)

    return pacientes_procesados
```

```python
def ordenar_pacientes(pacientes_procesados):

    pacientes_ordenados = sorted(
        pacientes_procesados,
        key=lambda paciente: paciente["riesgo_total"],
        reverse=True
    )

    return pacientes_ordenados
```

```python
def mostrar_resultados(pacientes_ordenados):

    df = pd.DataFrame(pacientes_ordenados)

    print("PACIENTES ORDENADOS POR PRIORIDAD:")
    print(df)

    return df
```

```python
def main_pacientes():

    pacientes_procesados = procesar_pacientes(pacientes)

    pacientes_ordenados = ordenar_pacientes(
        pacientes_procesados
    )

    mostrar_resultados(pacientes_ordenados)
```

```python
Nivell 3
Exercici 1: Generador de contrasenyes segures
Ets analista de dades dins el departament de Ciberseguretat d'una empresa tecnològica. S'ha detectat que la majoria de treballadors continuen utilitzant contrasenyes molt dèbils com “Asdf1234”,
dates d'aniversari o combinacions fàcils d'endevinar. Per aquest motiu, el responsable del departament t'ha demanat desenvolupar un generador de contrasenyes segures, totalment parametritzable i modular.
```

```python
import random
import string
```

```python
def crear_conjunto_caracteres(
    mayusculas,
    minusculas,
    numeros,
    signos
):

    conjuntos = []

    if mayusculas:
        conjuntos.append(string.ascii_uppercase)

    if minusculas:
        conjuntos.append(string.ascii_lowercase)

    if numeros:
        conjuntos.append(string.digits)

    if signos:
        conjuntos.append(string.punctuation)

    return conjuntos
```

```python
def asegurar_caracteres_obligatorios(conjuntos):

    caracteres_obligatorios = []

    for conjunto in conjuntos:

        caracter = random.choice(conjunto)

        caracteres_obligatorios.append(caracter)

    return caracteres_obligatorios
```

```python
def completar_contrasena(
    caracteres_obligatorios,
    conjuntos,
    longitud
):

    contrasena = caracteres_obligatorios.copy()

    todos_caracteres = "".join(conjuntos)

    while len(contrasena) < longitud:

        caracter = random.choice(todos_caracteres)

        contrasena.append(caracter)

    return contrasena
```

```python
def mezclar_contrasena(contrasena):

    random.shuffle(contrasena)

    contrasena_final = "".join(contrasena)

    return contrasena_final
```

```python
def crear_contrasena(
    longitud,
    mayusculas=True,
    minusculas=True,
    numeros=True,
    signos=False
):

    conjuntos = crear_conjunto_caracteres(
        mayusculas,
        minusculas,
        numeros,
        signos
    )

    if not validar_longitud(longitud, conjuntos):
        return "Error: longitud insuficiente"

    caracteres_obligatorios = asegurar_caracteres_obligatorios(
        conjuntos
    )

    contrasena = completar_contrasena(
        caracteres_obligatorios,
        conjuntos,
        longitud
    )

    contrasena_final = mezclar_contrasena(
        contrasena
    )

    return contrasena_final
```

```python
def validar_longitud(longitud, conjuntos):

    cantidad_tipos = len(conjuntos)

    if longitud < cantidad_tipos:
        return False

    return True
```

```python
Exercici 2: Processament automàtic de dades esportives
Ara formes part d'un equip d'analítica dins una organització esportiva que gestiona dades de la lliga catalana.
La teva companya de feina vol automatitzar la lectura del seu arxiu històric de resultats i extreure les estadístiques més importants.
T'ha demanat fer un petit sistema en Python que processi el fitxer 📄 "historic partits.txt"
🎯 Objectius
Mitjançant un codi modular, clar i documentat, el teu programa ha de calcular:

» 1.Quin equip és el més golejador.

» 2.Quin és el més golejat.

» 3.El nombre total de gols que ha fet cada equip.

» 4.La classificació final ordenada per punts on cada victòria suma 3 pts, cada empat 1 pts y cada derrota 0 pts.
```

```python
import pandas as pd
```

```python
def leer_archivo(nombre_archivo):

    with open(nombre_archivo, "r", encoding="utf-8") as archivo:

        lineas = archivo.readlines()

    return lineas
```

```python
def procesar_linea(linea):

    datos = linea.strip().split("\t")

    equipo_local = datos[0]
    resultado = datos[1]
    equipo_visitante = datos[2]

    goles = resultado.split("-")

    goles_local = int(goles[0])
    goles_visitante = int(goles[1])

    return equipo_local, goles_local, goles_visitante, equipo_visitante
```

```python
def inicializar_equipo(estadisticas, equipo):

    if equipo not in estadisticas:

        estadisticas[equipo] = {
            "goles_favor": 0,
            "goles_contra": 0,
            "puntos": 0
        }
```

```python
def actualizar_estadisticas(
    estadisticas,
    equipo_local,
    goles_local,
    goles_visitante,
    equipo_visitante
):

    estadisticas[equipo_local]["goles_favor"] += goles_local
    estadisticas[equipo_local]["goles_contra"] += goles_visitante

    estadisticas[equipo_visitante]["goles_favor"] += goles_visitante
    estadisticas[equipo_visitante]["goles_contra"] += goles_local

    if goles_local > goles_visitante:

        estadisticas[equipo_local]["puntos"] += 3

    elif goles_local < goles_visitante:

        estadisticas[equipo_visitante]["puntos"] += 3

    else:

        estadisticas[equipo_local]["puntos"] += 1
        estadisticas[equipo_visitante]["puntos"] += 1
```

```python
def procesar_partidos(lineas):

    estadisticas = {}

    for linea in lineas:

        equipo_local, goles_local, goles_visitante, equipo_visitante = procesar_linea(
            linea
        )

        inicializar_equipo(
            estadisticas,
            equipo_local
        )

        inicializar_equipo(
            estadisticas,
            equipo_visitante
        )

        actualizar_estadisticas(
            estadisticas,
            equipo_local,
            goles_local,
            goles_visitante,
            equipo_visitante
        )

    return estadisticas
```

```python
def crear_dataframe(estadisticas):

    df = pd.DataFrame.from_dict(
        estadisticas,
        orient="index"
    )

    df = df.reset_index()

    df = df.rename(
        columns={"index": "equipo"}
    )

    return df
```

```python
def ordenar_clasificacion(df):

    df_ordenado = df.sort_values(
        by="puntos",
        ascending=False
    )

    df_ordenado = df_ordenado.reset_index(
        drop=True
    )

    return df_ordenado
```

```python
def obtener_destacados(df):

    equipo_mas_goleador = df.loc[
        df["goles_favor"].idxmax(),
        "equipo"
    ]

    equipo_mas_goleado = df.loc[
        df["goles_contra"].idxmax(),
        "equipo"
    ]

    return equipo_mas_goleador, equipo_mas_goleado
```

```python
def mostrar_resultados(
    df_clasificacion,
    equipo_mas_goleador,
    equipo_mas_goleado
):

    print("CLASIFICACIÓN FINAL:")
    print(df_clasificacion)

    print(
        "\nEquipo más goleador:",
        equipo_mas_goleador
    )

    print(
        "Equipo más goleado:",
        equipo_mas_goleado
    )
```

```python
def main_deportes():

    lineas = leer_archivo("historic partits.txt")

    estadisticas = procesar_partidos(lineas)

    df = crear_dataframe(estadisticas)

    df_clasificacion = ordenar_clasificacion(df)

    equipo_mas_goleador, equipo_mas_goleado = obtener_destacados(
        df_clasificacion
    )

    mostrar_resultados(
        df_clasificacion,
        equipo_mas_goleador,
        equipo_mas_goleado
    )
```

```python
main_deportes()
```
