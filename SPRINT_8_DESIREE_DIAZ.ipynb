```python
# Crear la función para cargar el Excel
def cargar_datos(archivo):
    df = pd.read_excel(archivo, header=3)
    return df
```

```python
# Cargar el archivo Excel
df = cargar_datos("sample_data/sprint8_complex.xlsx")

# Mostrar las primeras cinco filas // tienen caracteres incorrectos
display(df.head())
```

```python
# Función para corregir caracteres extraños

def corregir_caracteres(df):

    df_corregido = df.copy()

    def corregir_texto(texto):
        if isinstance(texto, str):
            try:
                return texto.encode("latin1").decode("utf-8")
            except (UnicodeError, ValueError):
                return texto

        return texto

    df_corregido.columns = [
        corregir_texto(columna)
        for columna in df_corregido.columns
    ]

    for columna in df_corregido.select_dtypes(include="object").columns:
        df_corregido[columna] = df_corregido[columna].map(corregir_texto)

    return df_corregido
```

```python
# Función para corregir el símbolo del euro

def corregir_euros(df):

    df_corregido = df.copy()

    df_corregido["Salari mensual"] = (
        df_corregido["Salari mensual"]
        .str.replace("â‚¬", "€", regex=False)
        .str.replace("â‚", "€", regex=False)
    )

    return df_corregido
```

```python
# Crear una copia con los caracteres corregidos
df_corregido = corregir_caracteres(df)

# Corregir los símbolos del euro
df_corregido = corregir_euros(df_corregido)

# Mostrar una única tabla con todas las correcciones
display(df_corregido.head())
```

```python
# Función para analizar las dimensiones
def analizar_dimensiones(df):
    filas = df.shape[0]
    columnas = df.shape[1]

    print("Número de filas:", filas)
    print("Número de columnas:", columnas)
```

```python
# Mostramos los nombres de las columnas, para corregir los datos de codificacion y acentuacion
# FUNCIÓN: CORREGIR ERRORES DE CODIFICACIÓN

def corregir_texto(texto):

    if isinstance(texto, str):
        for codificacion in ["cp1252", "latin1"]:
            try:
                return texto.encode(codificacion).decode("utf-8")
            except (UnicodeError, ValueError):
                continue

    return texto


# Corregir nombres de columnas
df_limpio.columns = [
    corregir_texto(columna) for columna in df_limpio.columns
]

# Corregir los valores de texto
for columna in df_limpio.select_dtypes(include="object").columns:
    df_limpio[columna] = df_limpio[columna].map(corregir_texto)

# Comprobar resultados
display(df_limpio[["Nom", "Cognoms", "Salari mensual"]].head(20))
```

```python
# CORREGIR Y UNIFICAR LOS FORMATOS DE SALARIO

def corregir_salario(valor):

    if pd.isna(valor):
        return None

    valor = str(valor).strip().lower()

    # Corregir símbolos mal codificados
    valor = valor.replace("â‚¬", "€")

    # Convertir salarios escritos con palabras
    if valor == "mil vuit-cents":
        return 1800.0

    # Eliminar símbolos y palabras
    valor = valor.replace("euros", "")
    valor = valor.replace("€", "")
    valor = valor.replace(",", "")
    valor = valor.replace(" ", "")

    # Convertir a número
    return pd.to_numeric(valor, errors="coerce")


# Aplicamos la corrección de salarios
df_limpio["Salari mensual"] = (
    df_limpio["Salari mensual"].apply(corregir_salario)
)

# Comprobamos los resultados con el símbolo € y sin decimales
display(
    df_limpio[["Nom", "Cognoms", "Salari mensual"]]
    .head(20)
    .style.format({"Salari mensual": "{:.0f} €"})
)
```

```python
# ELIMINAR DUPLICADOS EXACTOS

filas_antes = len(df_limpio)

columnas = df_limpio.columns.drop("Unnamed: 0")

df_limpio = df_limpio.drop_duplicates(
    subset=columnas,
    keep="first"
).copy()

print("Registros eliminados:", filas_antes - len(df_limpio))
print("Registros restantes:", len(df_limpio))
```

```python
Partiremos del dataset validado en el ejercicio 2. Primero normalizaremos las categorías que representan el mismo concepto. Después comprobaremos la coherencia de la información sobre vehículos e hijos y unificaremos las columnas redundantes. Finalmente completaremos los valores ausentes de temperatura y salario mediante criterios estadísticos justificados, conservando todos los registros.
```

```python
# SPRINT 8 - NIVEL 1 - EJERCICIO 3
# NORMALIZAR VALORES CATEGÓRICOS

df_transformado = df_validado.copy()

# Normalizar género
df_transformado["Gènere"] = df_transformado["Gènere"].replace({
    "D": "Dona",
    "F": "Dona",
    "H": "Home",
    "M": "Home"
})
# Normalizar cargo
df_transformado["Càrrec"] = (
    df_transformado["Càrrec"]
    .str.strip()
    .str.lower()
    .str.title()
)

# Mostrar resultados
print("Géneros:")
print(df_transformado["Gènere"].value_counts(dropna=False))

print("\nCargos:")
print(df_transformado["Càrrec"].value_counts(dropna=False))
```

```python
# PASO 2 - IDENTIFICAR COLUMNAS DE VEHÍCULOS

print(df_transformado.columns.tolist())
```

```python
# PASO 2 - COMPROBAR DATOS DE VEHÍCULOS

# Ver los valores de la columna coche
print(df_transformado["Te_cotxe"].value_counts(dropna=False))

# Personas con kilómetros o consumo, pero sin coche declarado
incoherencias_coche = df_transformado[
    (df_transformado["Km_anuals"].notna() |
     df_transformado["Consum_mitja_L_100km"].notna()) &
    (df_transformado["Te_cotxe"].isna())
]

# Personas con consumo, pero sin kilómetros
sin_kilometros = df_transformado[
    df_transformado["Consum_mitja_L_100km"].notna() &
    df_transformado["Km_anuals"].isna()
]

print("\nSin coche declarado, pero con datos de vehículo:", len(incoherencias_coche))
print("Con consumo, pero sin kilómetros:", len(sin_kilometros))


#no modificamos los datos. Primero identificamos las incoherencias para decidir cómo corregirlas sin inventar kilómetros ni consumos.
#conservar los dos registros y dejar los kilómetros como NaN, porque no podemos calcularlos a partir del consumo medio.
```

```python
# Comprobar personas con datos de vehículo que declaran no tener coche

sin_coche = df_transformado[
    (df_transformado["Te_cotxe"] == False) &
    (
        df_transformado["Km_anuals"].notna() |
        df_transformado["Consum_mitja_L_100km"].notna()
    )
]

print("Personas sin coche, pero con kilómetros o consumo:", len(sin_coche))
display(sin_coche[["Te_cotxe", "Km_anuals", "Consum_mitja_L_100km"]])
```

```python
# Personas sin coche: kilómetros y consumo no aplicables

df_transformado.loc[
    df_transformado["Te_cotxe"] == False,
    ["Km_anuals", "Consum_mitja_L_100km"]
] = float("nan")

# Comprobar los dos casos con consumo, pero sin kilómetros
display(sin_kilometros[
    ["Te_cotxe", "Km_anuals", "Consum_mitja_L_100km"]
])
```

```python

#se normalizan los datos de personas sin coche y se mantienen los dos registros con consumo pero sin kilometraje, ya que no existe información suficiente para completar ese valor.
# mantenemos Km_anuals = NaN porque no podemos calcular los kilómetros sin información adicional.
```

```python
# PASO 3 - COMPROBAR INFORMACIÓN SOBRE HIJOS

print("Fills:")
print(df_transformado["Fills"].value_counts(dropna=False))

print("\nNo Fills:")
print(df_transformado["No Fills"].value_counts(dropna=False))

print("\nNombre_fills:")
print(df_transformado["Nombre_fills"].value_counts(dropna=False))
```

```python
# Comprobar contradicciones en la información de hijos

incoherencias_hijos = df_transformado[
    ((df_transformado["Fills"] == True) &
     (df_transformado["Nombre_fills"] == 0)) |
    ((df_transformado["No Fills"] == True) &
     (df_transformado["Nombre_fills"] > 0))
]

print("Registros con contradicciones:", len(incoherencias_hijos))

display(incoherencias_hijos[
    ["Fills", "No Fills", "Nombre_fills"]
])
```

```python
# Completar el número de hijos cuando falte

df_transformado.loc[
    df_transformado["Nombre_fills"].isna() &
    (df_transformado["No Fills"] == True) &
    (df_transformado["Fills"] != True),
    "Nombre_fills"
] = 0

# Conservar solo la columna del número de hijos
df_transformado = df_transformado.drop(
    columns=["Fills", "No Fills"]
)

# Comprobar resultado
print(df_transformado["Nombre_fills"].value_counts(dropna=False))
print("Registros totales:", len(df_transformado))
```

```python
conservamos el número declarado cuando existe. Si no existe y la persona indica únicamente que no tiene hijos, asignamos 0.
Si sabemos que tiene hijos, pero no cuántos, mantenemos NaN porque no podemos inventar la cantidad.
```

```python
# PASO 4 - COMPROBAR TEMPERATURAS

print("Temperaturas nulas:",
      df_transformado["Temperatura_mitjana_ciutat"].isna().sum())

print("\nTipo de dato:")
print(df_transformado["Temperatura_mitjana_ciutat"].dtype)

print("\nTemperaturas por ciudad:")
display(
    df_transformado.groupby("Ciutat")["Temperatura_mitjana_ciutat"]
    .agg(["count", "median"])
)
```

```python
# PASO 4 - IMPUTAR TEMPERATURAS POR CIUDAD

df_transformado["Temperatura_mitjana_ciutat"] = (
    df_transformado["Temperatura_mitjana_ciutat"].fillna(
        df_transformado.groupby("Ciutat")["Temperatura_mitjana_ciutat"]
        .transform("median")
    )
)

# Comprobar resultado
print("Temperaturas nulas restantes:",
      df_transformado["Temperatura_mitjana_ciutat"].isna().sum())

print("Registros totales:", len(df_transformado))
```

```python
se imputan las temperaturas ausentes utilizando la mediana de cada ciudad, porque representa mejor las condiciones climáticas locales y reduce la influencia de valores extremos.
```

```python
# PASO 5 - REVISAR SALARIOS

print("Salarios nulos:",
      df_transformado["Salari mensual"].isna().sum())

print("\nTipo de dato:")
print(df_transformado["Salari mensual"].dtype)

print("\nSalario mediano por cargo:")
display(
    df_transformado.groupby("Càrrec")["Salari mensual"]
    .agg(["count", "median"])
)
```

```python
# PASO 5 - COMPLETAR SALARIOS POR CARGO

df_transformado["Salari mensual"] = (
    df_transformado["Salari mensual"].fillna(
        df_transformado.groupby("Càrrec")["Salari mensual"]
        .transform("median")
    )
)

# Comprobar resultado
print("Salarios nulos restantes:",
      df_transformado["Salari mensual"].isna().sum())

print("Registros totales:", len(df_transformado))
```

```python
se imputan los 27 salarios ausentes mediante la mediana salarial del cargo profesional, conservando los salarios originales y evitando que valores extremos influyan excesivamente en la imputación.
```

```python
# ==========================================================
# PASO 5 - CALCULAR EDADES Y FRANJAS DE EDAD / podemos calcular el ajuste por edad para 908 empleados. Los otros 77 conservarán el ajuste como desconocido.
# ==========================================================

# 1. Guardar las fechas de nacimiento
df_enriquecido["Fecha_nacimiento"] = fechas

# 2. Calcular la edad a fecha 18/10/2026
fecha_referencia = pd.Timestamp("2026-10-18")

df_enriquecido["Edad"] = (
    fecha_referencia.year
    - df_enriquecido["Fecha_nacimiento"].dt.year
    - (
        fecha_referencia.strftime("%m%d")
        < df_enriquecido["Fecha_nacimiento"].dt.strftime("%m%d")
    )
)

# 3. Descartar edades imposibles o inferiores a 18 años
df_enriquecido.loc[
    (df_enriquecido["Edad"] < 18) |
    (df_enriquecido["Edad"] > 100),
    "Edad"
] = float("nan")

# 4. Asignar franjas de edad
def clasificar_edad(edad):

    if pd.isna(edad):
        return None
    elif edad < 30:
        return "18-29"
    elif edad < 45:
        return "30-44"
    elif edad < 60:
        return "45-59"
    else:
        return "60+"

df_enriquecido["Franja_edad"] = (
    df_enriquecido["Edad"].apply(clasificar_edad)
)

# 5. Mostrar resultados
print("EMPLEADOS POR FRANJA DE EDAD:")
print(df_enriquecido["Franja_edad"].value_counts(dropna=False))

display(
    df_enriquecido[
        ["Nom", "Fecha_nacimiento", "Edad", "Franja_edad"]
    ].head(10)
)
```

```python
# ==========================================================
# PASO 6 - APLICAR INCREMENTOS SALARIALES
# ==========================================================

# 1. Incremento por grupo profesional
df_enriquecido["Incremento_grupo"] = (
    df_enriquecido["Grup Professional"].map(
        politica["base_increment_by_group"]
    )
)

# 2. Incremento por hijos
def incremento_hijos(numero_hijos):

    if pd.isna(numero_hijos):
        return None
    elif numero_hijos > 0:
        return politica["children_adjustment"]["has_children"]
    else:
        return politica["children_adjustment"]["no_children"]

df_enriquecido["Incremento_hijos"] = (
    df_enriquecido["Nombre_fills"].apply(incremento_hijos)
)

# 3. Incremento por edad
df_enriquecido["Incremento_edad"] = (
    df_enriquecido["Franja_edad"].map(
        politica["age_adjustment"]
    )
)

# 4. Comprobar resultados
print("VALORES VACÍOS EN LOS INCREMENTOS:")
print(
    df_enriquecido[
        ["Incremento_grupo", "Incremento_hijos", "Incremento_edad"]
    ].isna().sum()
)

display(
    df_enriquecido[
        ["Nom", "Grup Professional", "Incremento_grupo",
         "Nombre_fills", "Incremento_hijos",
         "Franja_edad", "Incremento_edad"]
    ].head(10)
)
```

```python
# ==========================================================
# PASO 7.1 - ANALIZAR EL ÍNDICE DE COSTE DE VIDA
# ==========================================================

# 1. Estadísticas del índice
print("ESTADÍSTICAS DEL COSTE DE VIDA:")
print(df_ciudades["Cost_vida_index"].describe())

# 2. Mostrar ciudades e índices
print("\nCOSTE DE VIDA POR CIUDAD:")

display(
    df_ciudades[
        ["Ciutat", "País", "Cost_vida_index"]
    ].sort_values("Cost_vida_index")
)
```

```python
# PASO 7.2 - CLASIFICAR EL COSTE DE VIDA

def clasificar_coste(indice):

    if pd.isna(indice):
        return None
    elif indice < 68:
        return "low"
    elif indice <= 78:
        return "medium"
    else:
        return "high"

# Crear categoría de coste de vida
df_enriquecido["Categoria_coste"] = (
    df_enriquecido["Cost_vida_index"].apply(clasificar_coste)
)

# Aplicar incrementos del JSON
df_enriquecido["Incremento_coste"] = (
    df_enriquecido["Categoria_coste"].map(
        politica["cost_of_living_adjustment"]
    )
)

# Comprobar resultados
print("EMPLEADOS POR CATEGORÍA:")
print(df_enriquecido["Categoria_coste"].value_counts(dropna=False))

print("\nINCREMENTOS SIN CALCULAR:")
print(df_enriquecido["Incremento_coste"].isna().sum())
```

```python
# ==========================================================
# PASO 8 - CALCULAR EL SALARIO AJUSTADO
# ==========================================================

# 1. Seleccionar los cuatro incrementos
columnas_incrementos = [
    "Incremento_grupo",
    "Incremento_hijos",
    "Incremento_edad",
    "Incremento_coste"
]

# 2. Calcular el incremento total
df_enriquecido["Incremento_total"] = (
    df_enriquecido[columnas_incrementos].sum(
        axis=1, min_count=4
    )
)

# 3. Aplicar el límite máximo del 15 %
limite = politica["rules"]["max_total_increment_percent"]

df_enriquecido["Incremento_total"] = (
    df_enriquecido["Incremento_total"].clip(upper=limite)
)

# 4. Calcular el salario ajustado
df_enriquecido["Salario_ajustado"] = (
    df_enriquecido["Salari mensual"] *
    (1 + df_enriquecido["Incremento_total"] / 100)
)

# 5. Mostrar resultados
print("SALARIOS CALCULADOS:")
print(df_enriquecido["Salario_ajustado"].notna().sum())

print("\nSALARIOS PENDIENTES:")
print(df_enriquecido["Salario_ajustado"].isna().sum())

display(
    df_enriquecido[
        ["Nom", "Salari mensual",
         "Incremento_total", "Salario_ajustado"]
    ].head(10).style.format({
        "Salari mensual": "{:,.2f} €",
        "Incremento_total": "{:.2f} %",
        "Salario_ajustado": "{:,.2f} €"
    }, na_rep="—")
)
```

```python
# ==========================================================
# SPRINT 8 - NIVEL 2 - EJERCICIO 3
# PASO 2 - ÍNDICE DE DEPENDENCIA FAMILIAR
# ==========================================================

# Crear una copia del dataset original
df_derivadas = df_enriquecido.copy()

# Crear el índice de dependencia familiar
# Sumamos 1 para evitar dividir entre cero cuando no tiene hijos
df_derivadas["Indice_dependencia_familiar"] = (
    df_derivadas["Salari mensual"] /
    (df_derivadas["Nombre_fills"] + 1)
)

# Mostrar resultados con formato de euros
display(
    df_derivadas[
        ["Nom", "Salari mensual", "Nombre_fills",
         "Indice_dependencia_familiar"]
    ].head(10).style.format({
        "Salari mensual": "{:,.0f} €",
        "Nombre_fills": "{:.0f}",
        "Indice_dependencia_familiar": "{:,.2f} €"
    }, na_rep="—")
)
```

```python
# ==========================================================
# SPRINT 8 - NIVEL 2 - EJERCICIO 3
# PASO 3 - COMPROBAR CASOS LÍMITE
# ==========================================================

# Empleados sin hijos
print("EMPLEADOS SIN HIJOS:")
print((df_derivadas["Nombre_fills"] == 0).sum())

# Empleados sin información sobre hijos
print("\nVALORES NULOS:")
print(df_derivadas["Nombre_fills"].isna().sum())

# Empleados con número de hijos negativo
print("\nVALORES NEGATIVOS:")
print((df_derivadas["Nombre_fills"] < 0).sum())

# Resultados nulos del índice
print("\nÍNDICES NO CALCULADOS:")
print(df_derivadas["Indice_dependencia_familiar"].isna().sum())
```

```python
# ==========================================================
# SPRINT 8 - NIVEL 2 - EJERCICIO 3
# PASO 4 - RATIO SALARIO / COSTE DE VIDA
# Comparar salarios según el coste de vida de cada ciudad
# ==========================================================

# Crear la segunda variable derivada
df_derivadas["Ratio_salario_coste_vida"] = (
    df_derivadas["Salari mensual"] /
    df_derivadas["Cost_vida_index"].replace(0, float("nan"))
)

# Mostrar resultados con formato
display(
    df_derivadas[
        ["Nom", "Ciutat", "Salari mensual",
         "Cost_vida_index", "Ratio_salario_coste_vida"]
    ].head(10).style.format({
        "Salari mensual": "{:,.0f} €",
        "Cost_vida_index": "{:.0f}",
        "Ratio_salario_coste_vida": "{:.2f}"
    }, na_rep="—")
)
```

```python
 Utilitzant un bucle, exportar en 4 fitxers (format .csv o .xlsx) les dades corresponents a cada Grup Professional. Per exemple: dades_GrupA.csv, dades_GrupB.csv...
» Crear i exportar un cinquè DataFrame (format .csv o .xlsx) que contingui, per a cada Grup Professional:
Nombre total de treballadors.
Salari mitjà.
Edat mediana.
```

```python
# ==========================================================
# SPRINT 8 - NIVEL 2 - EJERCICIO 4
# PASO 1 - COMPROBAR GRUPOS PROFESIONALES
# ==========================================================

# Mostrar los grupos profesionales existentes
print("GRUPOS PROFESIONALES:")
print(df_derivadas["Grup Professional"].unique())

# Contar trabajadores por grupo
print("\nNÚMERO DE TRABAJADORES POR GRUPO:")
print(df_derivadas["Grup Professional"].value_counts())

# Comprobar las dimensiones del dataset
print("\nDIMENSIONES DEL DATASET:")
print(df_derivadas.shape)
```

```python
# ==========================================================
# SPRINT 8 - NIVEL 2 - EJERCICIO 4
# PASO 2 - EXPORTAR GRUPOS PROFESIONALES A CSV
# ==========================================================

# Lista de grupos profesionales
grupos = ["Grup A", "Grup B", "Grup C", "Grup D"]

# Recorrer los grupos con un bucle
for grupo in grupos:

    # Filtrar los trabajadores de cada grupo
    df_grupo = df_derivadas[
        df_derivadas["Grup Professional"] == grupo
    ]

    # Crear el nombre del archivo
    nombre_archivo = "datos_" + grupo.replace(" ", "") + ".csv"

    # Exportar el archivo CSV
    df_grupo.to_csv(nombre_archivo, index=False, encoding="utf-8-sig")

    # Mostrar confirmación
    print(nombre_archivo, ":", len(df_grupo), "trabajadores")
```

```python
# ==========================================================
# SPRINT 8 - NIVEL 2 - EJERCICIO 4
# PASO 3 - CREAR Y EXPORTAR EL RESUMEN
# ==========================================================

# Agrupar los datos por grupo profesional
df_resumen = df_derivadas.groupby("Grup Professional").agg(
    Total_trabajadores=("DNI", "count"),
    Salario_medio=("Salari mensual", "mean"),
    Edad_mediana=("Edad", "median")
).reset_index()

# Redondear el salario medio a dos decimales / Media: suma todos los valores y los divide entre el número de observaciones válidas.
df_resumen["Salario_medio"] = df_resumen["Salario_medio"].round(2)

# Exportar el quinto archivo CSV
df_resumen.to_csv(
    "resumen_grupos.csv",
    index=False,
    encoding="utf-8-sig"
)

# Mostrar el resultado con formato de euros / Mediana: es el valor central cuando ordenamos los datos de menor a mayor. Es menos sensible a valores extremos.
display(
    df_resumen.style.format({
        "Salario_medio": "{:,.2f} €",
        "Edad_mediana": "{:.0f}"
    }, na_rep="—")
)

print("Archivo creado: resumen_grupos.csv")
```

```python
# ==========================================================
# SPRINT 8 - NIVEL 2 - EJERCICIO 4
# PASO 4 - EXPORTAR TODO A UN EXCEL
# ==========================================================

# Crear un archivo Excel con varias pestañas
with pd.ExcelWriter("resultados_sprint8.xlsx") as writer:

    # Crear una pestaña para cada grupo profesional
    for grupo in grupos:

        df_grupo = df_derivadas[
            df_derivadas["Grup Professional"] == grupo
        ]

        # Nombre de la pestaña
        nombre_hoja = grupo.replace(" ", "")

        # Guardar el grupo en su pestaña
        df_grupo.to_excel(
            writer,
            sheet_name=nombre_hoja,
            index=False
        )

    # Crear la pestaña del resumen
    df_resumen.to_excel(
        writer,
        sheet_name="Resumen",
        index=False
    )

print("Archivo Excel creado: resultados_sprint8.xlsx")
```
